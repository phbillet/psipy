# Ray fans through a softened Coulomb singularity

**Goal**. Follow the wavefront set of a point source as it is moved along the $x$-axis through the centre of an attractive Coulomb-type potential, and see how the fan of rays depends on

- the **size $r$** of the source,
- the **energy** of the rays: **bound** rays ($p < 0$, covector norm $1$) versus **scattering** rays ($p > 0$, covector norm $5$),
- the **view** of the phase space: four 2D slices, or the cotangent bundle itself drawn as base plane + circle fibre.

---

## 1. The operator

We work in $\mathbb{R}^2$ with the symbol

$$
p(x,y,\xi,\eta) = \frac{\xi^{2}+\eta^{2}}{2m} - \frac{k}{\sqrt{x^{2}+y^{2}+\varepsilon}} .
$$

With the Kohn–Nirenberg quantization,

$$
(Pu)(x) = \frac{1}{(2\pi)^2} \iint e^{i x \cdot \xi} p(x,\xi) \hat{u}(\xi) \, d\xi
\quad \Longrightarrow \quad
P = -\frac{1}{2m}\Delta + V(x), \qquad V(x) = -\frac{k}{\sqrt{|x|^{2}+\varepsilon}} \qquad (\hbar=1),
$$

so $P$ is a Schrödinger operator with a **softened Coulomb potential**. Only the symbol `op.symbol` is used below.

| Parameter | Value | Meaning |
|---|---|---|
| $m$ | $1$ | mass |
| $k$ | $1$ | coupling (attractive) |
| $\varepsilon$ | $0.01$ | softening |

The singularity of the pure Coulomb symbol ($\varepsilon=0$, $V \sim -k/|x|$) at the origin is regularised by $\varepsilon$. The regularised potential has

- core radius $\sqrt{\varepsilon} = 0.1$,
- depth $V(0) = -k/\sqrt{\varepsilon} = -10$,
- maximal force $F_{\max} = \frac{2k}{3\sqrt{3}\,\varepsilon} \approx 38.5$, reached at $|x| = \sqrt{\varepsilon/2} \approx 0.07$.

---

## 2. Rays = bicharacteristics

Singularities of a solution propagate along the Hamiltonian flow of $p$:

$$
\dot{x} = \frac{\partial p}{\partial \xi} = \frac{\xi}{m}, \qquad
\dot{y} = \frac{\partial p}{\partial \eta} = \frac{\eta}{m}, \qquad
\dot{\xi} = -\frac{\partial p}{\partial x} = -\frac{k\,x}{(x^{2}+y^{2}+\varepsilon)^{3/2}}, \qquad
\dot{\eta} = -\frac{\partial p}{\partial y} = -\frac{k\,y}{(x^{2}+y^{2}+\varepsilon)^{3/2}} .
$$

Two quantities are conserved along each ray: the symbol itself, $p$ (the **energy** $E$ of the ray), and the angular momentum $L = x\eta - y\xi$ (the symbol is rotation invariant). The impact parameter of a ray with respect to the centre is $b = L/|\kappa|$, where $|\kappa| = \sqrt{\xi^2+\eta^2}$ is its covector norm.

> This is the **geometric (semiclassical) picture**: the animation shows what the theory of propagation of singularities predicts. It is not a simulation of a wave equation (no amplitudes, no interference).

---

## 3. The point source and the energy

A point source at $(x_0,0)$ contains **every** direction of the covector. It is modelled by $N=300$ seeds on a small circle of radius $r$ around the source, each carrying the outward covector of norm $|\kappa| = \sqrt{\xi_0^2+\eta_0^2}$:

$$
(x_j,y_j) = \bigl(x_0+r\cos\theta_j,\; r\sin\theta_j\bigr), \qquad
(\xi_j,\eta_j) = |\kappa| \bigl(\cos\theta_j,\;\sin\theta_j\bigr), \qquad
\theta_j = \frac{2\pi j}{N}.
$$

If `xi0=0, eta0=0`, the norm takes its default value $|\kappa|=1$. The energy of a seed is

$$
p_j = \frac{|\kappa|^{2}}{2m} - \frac{k}{\sqrt{(x_0+r\cos\theta_j)^2 + r^{2}\sin^{2}\theta_j + \varepsilon}} ,
$$

and the angular momentum is $L = x_0 |\kappa| \sin\theta_j$, i.e. the impact parameter is $b = x_0 \sin\theta_j$.

Two consequences:

- **For a source at the centre** ($x_0=0$) all rays have the same energy $p_0(r) = \frac{|\kappa|^2}{2} - k/\sqrt{r^2+\varepsilon}$ and $L=0$: every ray is **radial** and passes through the core.
- **For a source at $x_0 \neq 0$** the energy is approximately $p \approx \frac{|\kappa|^{2}}{2} - k/\sqrt{x_0^2+\varepsilon}$. Only rays with $|b| \lesssim \sqrt{\varepsilon} = 0.1$ that head towards the centre cross the core.

### Two regimes

| Source at the centre | $r=1$ | $r=0.1$ | $r=0.01$ | $r=0.001$ |
|---|---|---|---|---|
| $p_0$ for $|\kappa|=1$ (`xi0=0`) | $-0.495$ | $-6.57$ | $-9.45$ | $-9.50$ |
| $p_0$ for $|\kappa|=5$ (`xi0=5`) | $11.5$ | $5.43$ | $2.55$ | $2.50$ |

- **Bound rays, $|\kappa|=1$.** At $x_0=0$ the energy is negative, so the rays cannot escape. They oscillate radially inside the turning radius $r_t$ defined by $V(r_t) = p_0$: $r_t \approx 2.0$ for $r=1$, and $r_t \approx 0.115,\; 0.035,\; 0.033$ for $r=0.1,\; 0.01,\; 0.001$, i.e. inside the core. As the source moves away from the centre the energy rises and **changes sign** at $|x_0| \approx 2.0$: beyond that the rays are unbound. The wide sweep $x_0 \in (-5,5)$ therefore crosses the threshold $p=0$.
- **Scattering rays, $|\kappa|=5$.** The energy is positive everywhere along the sweep (from about $2.5$ to $5.4$ at $x_0=0$, and about $12.3$ far away). The rays escape to infinity and are only deflected by the attractive centre. Positive energy at the centre requires $|\kappa| > \sqrt{2mk/\sqrt{\varepsilon}} \approx 4.47$.

Compare $r$ with the core radius $\sqrt{\varepsilon} = 0.1$:

- $r = 1 \gg \sqrt{\varepsilon}$: the seed circle surrounds the core;
- $r = 0.1 \approx \sqrt{\varepsilon}$: the circle has the size of the core;
- $r = 0.01,\, 0.001 \ll \sqrt{\varepsilon}$: the circle lies deep inside the core, close to the ideal point source ($r \to 0$).

## 4. The spatial choreography of the cotangent bundle: the $(x,y,\varphi)$ view

### 4.1 Phase space as a kinetic bundle

The phase space $T^*\mathbb{R}^2 = \mathbb{R}^2_{(x,y)} \times \mathbb{R}^2_{(\xi,\eta)}$ is a fibre bundle over the spatial base $\mathbb{R}^2_{(x,y)}$. Above each point in space lies a fibre—the plane of covectors (momenta) at that location. By passing to polar coordinates in the fibre,

$$
\xi = \kappa\cos\varphi, \qquad \eta = \kappa\sin\varphi,
$$

we separate the **kinetic energy scale** $\kappa = \sqrt{\xi^2+\eta^2}$ from the **directional phase** $\varphi$. The wavefront set of a point source at $(x_0,0)$ is the entire punctured fibre above $x_0$. Our seeds form the circle $\kappa = |\kappa_0|$ within this fibre, with $\varphi = \theta_j$, displaced by the physical radius $r$.

### 4.2 The 3D stage and visual tension

Visualizing 4D phase space requires resolving visual tension. With `projection='3d'`, `coords3d=('x','y','phi')`, and `color_by='kappa'`, the plot maps:

- the **base** $(x,y)$ as the horizontal spatial stage,
- the **covector angle** $\varphi$ as the vertical axis, representing the unrolled circle fibre,
- the **covector norm** $\kappa$ as the colour, encoding the kinetic energy.

At $t=0$, the green initial points form a vertical line of height $2\pi$ above the source—the circle fibre cut open and unrolled to $\varphi \in [-\pi, \pi)$. The Hamiltonian flow carries this line into a 2D surface. Two camera perspectives reveal different aspects of this choreography:

- The **top view** (`view=(90,-90)`) projects onto the base; the folds of this surface are the *caustics*, where neighbouring rays cross in space.
- The **side view** (`view=(0,-90)`) reveals the momentum winding, showing how the covector turns along the wavefront.

`ortho=True` gives an orthographic camera, and `rotate=2` turns the camera slowly during the animation.

### 4.3 The rhythm of $\varphi$ and $\kappa$: turning points and reversals

In the fibre's polar coordinates, the flow dictates the evolution of the angle:

$$
\dot{\varphi} = \frac{\xi\dot{\eta} - \eta\dot{\xi}}{\kappa^{2}} = \frac{V'(\rho)}{\rho\,\kappa^{2}}\,L, \qquad \rho = \sqrt{x^2+y^2},
$$

where $L = x\eta - y\xi$ is the angular momentum. The covector turns **only** where $L \neq 0$. For a source at the centre ($L=0$), every ray preserves its angle $\varphi$; the surface is a perfect **helicoid**—a spiral ramp whose axis is the vertical line above the source. Away from the centre, the helicoid is bent by the spatial deflection of the rays.

Crucially, $\varphi$ is undefined where $\kappa = 0$. This occurs at the **turning points** of bound radial rays (source exactly at the centre, $L=0$): the covector norm vanishes and the momentum reverses, causing $\varphi$ to jump by $\pi$. This is a coordinate artefact of the polar chart, not a physical rotation. As soon as $L \neq 0$ (source slightly off-centre), the covector never vanishes; it turns continuously, winding several times as it passes the core.

## 6. Reading the wavefront choreography: what to look for

These expectations follow from the geometric optics equations; compare them with the visual rhythm of the animations.

- **Scattering rays (Parts A and C): The unbound trajectory.**
  There is no trapping. The fan expands and decelerates as it climbs out of the potential well. At $x_0=0$, all rays are radial ($L=0$), meaning there is no spatial deflection: $\varphi$ stays constant, $\kappa$ decreases from $5$ to about $3.4$, and the 3D surface is a pristine helicoid. As $x_0$ grows, rays with impact parameter $|b| < 0.1$ cross the core, while others are deflected. The visual tension increases: the covector angle changes, and the surface warps. Far from the centre, the choreography settles back into a helicoid around the moving source.

- **Bound rays (Parts B and C): The ritual of the turning points.**
  For $r \le 0.1$ and small $|x_0|$, the rays are trapped in the core and oscillate radially. In the $(x,y,\varphi)$ view, the covector reverses at each turning point, manifesting as the $\pi$-jump artefact described in Section 4.3. The accumulated angle can reach $6\pi$ within $t \in [0,1]$. The most striking transition occurs around $|x_0| \approx 2$: the fan switches from bound to unbound, and the visual structure changes abruptly. Furthermore, as soon as the source leaves the exact centre ($x_0 \neq 0$), the discrete $\pi$-reversals are replaced by continuous winding, altering the topology of the wavefront surface.

- **Dependence on the source radius $r$: The macroscopic vs microscopic scale.**
  The pictures of Part B should converge as $r \to 0$ (the ideal point source limit). The critical geometric transition happens around $r \approx \sqrt{\varepsilon} = 0.1$. When $r \gg \sqrt{\varepsilon}$, the seed circle surrounds the singular core; when $r \ll \sqrt{\varepsilon}$, the circle lies deep inside the regularised core, probing the steep potential gradient. In the $(x,y,\varphi)$ view, the radial bound rays ($x_0=0$) alternate between two heights at their turning points, whereas for $x_0 \neq 0$ the covector winds continuously, so the picture changes abruptly as soon as the source leaves the centre.

## 7. Numerical remark

The rays are integrated with a first-order fixed-step symplectic scheme (`n_steps=500` by default, step $0.002$). Near the core the force is large, so the energy $p$ is only conserved approximately. The largest energy drift over a ray was

| Rays | `n_steps` $=500$ | $2000$ | $4000$ | $8000$ |
|---|---|---|---|---|
| bound, $|\kappa|=1$ | $\approx 0.15$ | $\approx 0.04$ | n/a | $\approx 0.01$ |
| scattering, $|\kappa|=5$, $r=0.1$ | $0.18$–$0.26$ | n/a | $0.02$–$0.03$ | n/a |

These are absolute errors. For $|\kappa|=5$ they are about $2$–$3\%$ of the energy at `n_steps=500` and about $0.3$–$0.6\%$ at $4000$. For bound rays of energy around $-0.5$ they can be tens of percent at `n_steps=500`. Before drawing conclusions, check that a picture does not change when `n_steps` is increased (for example to `4000`).

In the 3D view each ray is drawn with at most about 200 points, so the rendering stays light even for large `n_steps`.

````python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import HTML

# Assuming psiop is on the python path
from psiop import *

# Spatial and frequency variables
x, y, xi, eta = sp.symbols('x y xi eta', real=True)

A = 1
B = 1
m = 1
k = 1
alpha = 1
beta = 1

eps = 0.01

symbol_expr = (xi**2 + eta**2)/(2*m) - k/sp.sqrt(x**2 + y**2 + eps)
# symbol_expr = sp.sqrt(xi**2 + eta**2 + m**2) - k/sp.sqrt(x**2 + y**2 + eps)

op = PseudoDifferentialOperator(
    expr=symbol_expr,
    vars_x=[x, y],
    mode='symbol',
    quantization='kohn-nirenberg',
    apply_backend='peetre',
)

print('Operator symbol:')
sp.pprint(op.symbol)
print(f'\ndimension           : {op.dim}')
print(f'quantization        : {op.quantization}')
print(f'apply backend       : {op.apply_backend}')
print(f'spatially dependent : {op._is_spatial_dependent()}')
````

## Diagnostic: following one ray

The function below follows **a single ray** of the point-source fan and reports how the covector norm $\kappa$ and the covector angle $\varphi$ evolve along it:

$$
\kappa=\sqrt{\xi^{2}+\eta^{2}},\qquad \varphi=\arg(\xi+i\eta).
$$

The ray starts at the seed of angle $\theta$ of the circle of Section 3,

$$
(x,y)=\bigl(x_0+r\cos\theta,\; r\sin\theta\bigr),\qquad
(\xi,\eta)=\kappa_0\bigl(\cos\theta,\;\sin\theta\bigr),
$$

and is integrated with the same symplectic scheme as the animations, here with `n_steps=8000`.

| Argument | Meaning |
|---|---|
| `x0`, `r` | centre and radius of the seed circle |
| `kappa` | initial covector norm $\kappa_0$ (`1`: bound rays, `5`: scattering rays) |
| `theta` | direction of the seed on the circle |
| `tspan`, `n_steps` | flow time and number of steps |
| `plot` | also draw $\kappa(t)$ and $\varphi(t)/\pi$ |

It prints the smallest value of $\kappa$ along the ray, the number of **jumps** of $\varphi$ (steps larger than $1$ rad, i.e. of size about $\pi$) with their sizes in units of $\pi$, and the net change $\varphi_{\text{final}}-\varphi_0$.

> The angle is computed here with the plain `np.unwrap`, which removes **every** jump larger than $\pi$. It therefore shows the raw behaviour of the arctangent. The $(x,y,\varphi)$ animations use a different unwrapping that removes only the $2\pi$ wraps and leaves the $\pi$ reversals as they are.

### The three calls

| Call | Ray |
|---|---|
| `theta=0.0`, `kappa=1.0` | bound, source at the centre, direction $\theta=0$ |
| `theta=1.0`, `kappa=1.0` | the same flow, another direction |
| `kappa=5.0` | scattering, source at the centre |

### Expected output

| Call | $\kappa_{\min}$ | jumps of $\varphi$ | $\varphi_{\text{final}}-\varphi_0$ |
|---|---|---|---|
| bound, $\theta=0$ | $0.0002$ | $6$, sizes $+1,-1,+1,-1,+1,-1$ (in units of $\pi$) | $0\,\pi$ |
| bound, $\theta=1$ | $0.0002$ | $6$, all $+1$ | $6\,\pi$ |
| scattering | $3.3741$ | none | $0\,\pi$ |

How to read it:

- **Bound rays** ($\lvert\kappa\rvert=1$, $L=0$). The ray is radial and oscillates through the core. Its energy is $p_0=-6.57$, so $\kappa$ oscillates between $0$ and $\kappa_{\max}=\sqrt{2\,(p_0+k/\sqrt\varepsilon)}\approx2.6$. The six zeros of $\kappa$ are the turning points, about $0.17$ apart in time, and the six jumps of $\varphi$ occur exactly there: the covector vanishes and reverses, so $\varphi$ changes by $\pi$. The two directions have the **same flow** (the same $\kappa(t)$), but the sign of the jumps is arbitrary: it alternates for $\theta=0$ (the angle returns to its start) and is always $+$ for $\theta=1$ (a staircase up to $6\pi$). This is an artefact of the angle coordinate, not of the flow.
- **Scattering ray** ($\lvert\kappa\rvert=5$, $L=0$). $\kappa$ decreases from $5$ to $3.37$ as the ray climbs out of the well and never vanishes. $\varphi$ stays exactly constant, which is why the surface of rays is a helicoid.

````python
def ray_diagnostics(op, x0=0.0, r=0.1, kappa=1.0, theta=0.0, tspan=(0, 1), n_steps=8000, plot=True):
    """Follow ONE ray of the point-source fan and report how kappa and phi evolve."""
    z0 = (x0 + r*np.cos(theta), r*np.sin(theta), kappa*np.cos(theta), kappa*np.sin(theta))
    tr = bicharacteristic_flow(op.symbol, z0, tspan, dim=2, method='symplectic', n_steps=n_steps)
    t, xi_, eta_ = np.asarray(tr['t']), np.asarray(tr['xi']), np.asarray(tr['eta'])
    kap = np.hypot(xi_, eta_)
    phi = np.unwrap(np.arctan2(eta_, xi_))
    steps = np.diff(phi)
    jumps = np.where(np.abs(steps) > 1.0)[0]                 # steps of about +-pi
    print(f"n_steps={n_steps}: kappa_min={kap.min():.4f}, {len(jumps)} jump(s) of phi, "
          f"sizes/pi = {np.round(steps[jumps]/np.pi, 2).tolist()}, "
          f"phi_final - phi_0 = {(phi[-1]-phi[0])/np.pi:.2f} pi")
    if plot:
        fig, ax = plt.subplots(1, 2, figsize=(10, 3.2))
        ax[0].plot(t, kap);       ax[0].set_xlabel('t'); ax[0].set_ylabel('kappa')
        ax[1].plot(t, phi/np.pi); ax[1].set_xlabel('t'); ax[1].set_ylabel('phi / pi')
        plt.tight_layout(); plt.show()

ray_diagnostics(op, kappa=1.0, theta=0.0)   # bound, source at the centre
ray_diagnostics(op, kappa=1.0, theta=1.0)   # same, another direction
ray_diagnostics(op, kappa=5.0)              # scattering, source at the centre
````

# In French

# Éventails de rayons à travers une singularité coulombienne adoucie

**Objectif.** Suivre l'ensemble de front d'onde (*wavefront set*) d'une source ponctuelle alors qu'elle se déplace le long de l'axe $x$ à travers le centre d'un potentiel attractif de type coulombien, et observer comment l'éventail de rayons dépend :
- de la **taille $r$** de la source,
- de l'**énergie** des rayons : rayons **liés** ($p < 0$, norme du covecteur $1$) par rapport aux rayons de **diffusion** ($p > 0$, norme du covecteur $5$),
- de la **vue** de l'espace des phases : quatre coupes 2D, ou le fibré cotangent lui-même représenté comme un plan de base + une fibre circulaire.

---

## 1. L'opérateur

Nous travaillons dans $\mathbb{R}^2$ avec le symbole

$$
p(x,y,\xi,\eta) = \frac{\xi^{2}+\eta^{2}}{2m} - \frac{k}{\sqrt{x^{2}+y^{2}+\varepsilon}} .
$$

Avec la quantification de Kohn-Nirenberg,

$$
(Pu)(x) = \frac{1}{(2\pi)^2} \iint e^{i x \cdot \xi} p(x,\xi) \hat{u}(\xi) \, d\xi
\quad \Longrightarrow \quad
P = -\frac{1}{2m}\Delta + V(x), \qquad V(x) = -\frac{k}{\sqrt{|x|^{2}+\varepsilon}} \qquad (\hbar=1),
$$

de sorte que $P$ est un opérateur de Schrödinger avec un **potentiel coulombien adouci**. Seul le symbole `op.symbol` est utilisé ci-dessous.

| Paramètre | Valeur | Signification |
|---|---|---|
| $m$ | $1$ | masse |
| $k$ | $1$ | couplage (attractif) |
| $\varepsilon$ | $0.01$ | adoucissement (régularisation) |

La singularité du symbole coulombien pur ($\varepsilon=0$, $V \sim -k/|x|$) à l'origine est régularisée par $\varepsilon$. Le potentiel régularisé possède :
- un rayon de cœur $\sqrt{\varepsilon} = 0.1$,
- une profondeur $V(0) = -k/\sqrt{\varepsilon} = -10$,
- une force maximale $F_{\max} = \frac{2k}{3\sqrt{3}\,\varepsilon} \approx 38.5$, atteinte à $|x| = \sqrt{\varepsilon/2} \approx 0.07$.

---

## 2. Rayons = bicharactéristiques

Les singularités d'une solution se propagent le long du flot hamiltonien de $p$ :

$$
\dot{x} = \frac{\partial p}{\partial \xi} = \frac{\xi}{m}, \qquad
\dot{y} = \frac{\partial p}{\partial \eta} = \frac{\eta}{m}, \qquad
\dot{\xi} = -\frac{\partial p}{\partial x} = -\frac{k\,x}{(x^{2}+y^{2}+\varepsilon)^{3/2}}, \qquad
\dot{\eta} = -\frac{\partial p}{\partial y} = -\frac{k\,y}{(x^{2}+y^{2}+\varepsilon)^{3/2}} .
$$

Deux grandeurs sont conservées le long de chaque rayon : le symbole lui-même, $p$ (l'**énergie** $E$ du rayon), et le moment angulaire $L = x\eta - y\xi$ (le symbole est invariant par rotation). Le paramètre d'impact d'un rayon par rapport au centre est $b = L/|\kappa|$, où $|\kappa| = \sqrt{\xi^2+\eta^2}$ est la norme de son covecteur.

> Il s'agit du **tableau géométrique (semi-classique)** : l'animation montre ce que prédit la théorie de la propagation des singularités. Ce n'est pas une simulation d'une équation d'onde (pas d'amplitudes, pas d'interférences).

---

## 3. La source ponctuelle et l'énergie

Une source ponctuelle en $(x_0,0)$ contient **toutes** les directions du covecteur. Elle est modélisée par $N=300$ graines sur un petit cercle de rayon $r$ autour de la source, chacune portant le covecteur sortant de norme $|\kappa| = \sqrt{\xi_0^2+\eta_0^2}$ :

$$
(x_j,y_j) = \bigl(x_0+r\cos\theta_j,\; r\sin\theta_j\bigr), \qquad
(\xi_j,\eta_j) = |\kappa| \bigl(\cos\theta_j,\;\sin\theta_j\bigr), \qquad
\theta_j = \frac{2\pi j}{N}.
$$

Si `xi0=0, eta0=0`, la norme prend sa valeur par défaut $|\kappa|=1$. L'énergie d'une graine est

$$
p_j = \frac{|\kappa|^{2}}{2m} - \frac{k}{\sqrt{(x_0+r\cos\theta_j)^2 + r^{2}\sin^{2}\theta_j + \varepsilon}} ,
$$

et le moment angulaire est $L = x_0 |\kappa| \sin\theta_j$, c'est-à-dire que le paramètre d'impact est $b = x_0 \sin\theta_j$.

Deux conséquences :
- **Pour une source au centre** ($x_0=0$), tous les rayons ont la même énergie $p_0(r) = \frac{|\kappa|^2}{2} - k/\sqrt{r^2+\varepsilon}$ et $L=0$ : chaque rayon est **radial** et traverse le cœur.
- **Pour une source en $x_0 \neq 0$**, l'énergie est approximativement $p \approx \frac{|\kappa|^{2}}{2} - k/\sqrt{x_0^2+\varepsilon}$. Seuls les rayons avec $|b| \lesssim \sqrt{\varepsilon} = 0.1$ qui se dirigent vers le centre traversent le cœur.

### Deux régimes

| Source au centre | $r=1$ | $r=0.1$ | $r=0.01$ | $r=0.001$ |
|---|---|---|---|---|
| $p_0$ pour $|\kappa|=1$ (`xi0=0`) | $-0.495$ | $-6.57$ | $-9.45$ | $-9.50$ |
| $p_0$ pour $|\kappa|=5$ (`xi0=5`) | $11.5$ | $5.43$ | $2.55$ | $2.50$ |

- **Rayons liés, $|\kappa|=1$.** En $x_0=0$, l'énergie est négative, les rayons ne peuvent donc pas s'échapper. Ils oscillent radialement à l'intérieur du rayon de rebroussement $r_t$ défini par $V(r_t) = p_0$ : $r_t \approx 2.0$ pour $r=1$, et $r_t \approx 0.115,\; 0.035,\; 0.033$ pour $r=0.1,\; 0.01,\; 0.001$, c'est-à-dire à l'intérieur du cœur. À mesure que la source s'éloigne du centre, l'énergie augmente et **change de signe** à $|x_0| \approx 2.0$ : au-delà, les rayons ne sont plus liés. Le balayage large $x_0 \in (-5,5)$ traverse donc le seuil $p=0$.
- **Rayons de diffusion, $|\kappa|=5$.** L'énergie est positive partout le long du balayage (d'environ $2.5$ à $5.4$ en $x_0=0$, et environ $12.3$ au loin). Les rayons s'échappent vers l'infini et sont seulement déviés par le centre attractif. Une énergie positive au centre requiert $|\kappa| > \sqrt{2mk/\sqrt{\varepsilon}} \approx 4.47$.

Comparons $r$ au rayon du cœur $\sqrt{\varepsilon} = 0.1$ :
- $r = 1 \gg \sqrt{\varepsilon}$ : le cercle des graines entoure le cœur ;
- $r = 0.1 \approx \sqrt{\varepsilon}$ : le cercle a la taille du cœur ;
- $r = 0.01,\, 0.001 \ll \sqrt{\varepsilon}$ : le cercle se trouve profondément à l'intérieur du cœur, proche de la source ponctuelle idéale ($r \to 0$).

## 4. La chorégraphie spatiale du fibré cotangent : la vue $(x,y,\varphi)$

### 4.1 L'espace des phases comme fibré cinétique

L'espace des phases $T^*\mathbb{R}^2 = \mathbb{R}^2_{(x,y)} \times \mathbb{R}^2_{(\xi,\eta)}$ est un fibré au-dessus de la base spatiale $\mathbb{R}^2_{(x,y)}$. Au-dessus de chaque point de l'espace se trouve une fibre : le plan des covecteurs (impulsions) à cet endroit. En passant aux coordonnées polaires dans la fibre,

$$
\xi = \kappa\cos\varphi, \qquad \eta = \kappa\sin\varphi,
$$

nous séparons l'**échelle d'énergie cinétique** $\kappa = \sqrt{\xi^2+\eta^2}$ de la **phase directionnelle** $\varphi$. L'ensemble de front d'onde d'une source ponctuelle en $(x_0,0)$ est la fibre entière (épointée) au-dessus de $x_0$. Nos graines forment le cercle $\kappa = |\kappa_0|$ au sein de cette fibre, avec $\varphi = \theta_j$, décalé par le rayon physique $r$.

### 4.2 La scène 3D et la tension visuelle

Visualiser un espace des phases en 4D exige de résoudre une tension visuelle. Avec `projection='3d'`, `coords3d=('x','y','phi')`, et `color_by='kappa'`, le graphique mappe :
- la **base** $(x,y)$ comme la scène spatiale horizontale,
- l'**angle du covecteur** $\varphi$ comme l'axe vertical, représentant le cercle de la fibre déroulé,
- la **norme du covecteur** $\kappa$ comme la couleur, encodant l'énergie cinétique.

À $t=0$, les points initiaux verts forment une ligne verticale de hauteur $2\pi$ au-dessus de la source : la fibre circulaire coupée et déroulée en $\varphi \in [-\pi, \pi)$. Le flot hamiltonien transporte cette ligne en une surface 2D. Deux perspectives de caméra révèlent différents aspects de cette chorégraphie :
- La **vue de dessus** (`view=(90,-90)`) projette sur la base ; les plis de cette surface sont les *caustiques*, là où les rayons voisins se croisent dans l'espace.
- La **vue de côté** (`view=(0,-90)`) révèle l'enroulement de l'impulsion, montrant comment le covecteur tourne le long du front d'onde.

`ortho=True` donne une caméra orthographique, et `rotate=2` fait tourner lentement la caméra pendant l'animation.

### 4.3 Le rythme de $\varphi$ et $\kappa$ : points de rebroussement et renversements

Dans les coordonnées polaires de la fibre, le flot dicte l'évolution de l'angle :

$$
\dot{\varphi} = \frac{\xi\dot{\eta} - \eta\dot{\xi}}{\kappa^{2}} = \frac{V'(\rho)}{\rho\,\kappa^{2}}\,L, \qquad \rho = \sqrt{x^2+y^2},
$$

où $L = x\eta - y\xi$ est le moment angulaire. Le covecteur ne tourne **que** là où $L \neq 0$. Pour une source au centre ($L=0$), chaque rayon préserve son angle $\varphi$ ; la surface est un parfait **hélicoïde** : une rampe en spirale dont l'axe est la ligne verticale au-dessus de la source. Loin du centre, l'hélicoïde est courbé par la déflexion spatiale des rayons.

De manière cruciale, $\varphi$ n'est pas défini là où $\kappa = 0$. Cela se produit aux **points de rebroussement** des rayons radiaux liés (source exactement au centre, $L=0$) : la norme du covecteur s'annule et l'impulsion s'inverse, provoquant un saut de $\varphi$ de $\pi$. Il s'agit d'un artefact de coordonnées du graphe polaire, et non d'une rotation physique. Dès que $L \neq 0$ (source légèrement décentrée), le covecteur ne s'annule jamais ; il tourne continûment, s'enroulant plusieurs fois en passant près du cœur.

## 6. Lire la chorégraphie du front d'onde : ce qu'il faut observer

Ces attentes découlent des équations de l'optique géométrique ; comparez-les avec le rythme visuel des animations.

- **Rayons de diffusion (Parties A et C) : La trajectoire non liée.**
  Il n'y a pas de piégeage. L'éventail s'étend et ralentit en sortant du puits de potentiel. En $x_0=0$, tous les rayons sont radiaux ($L=0$), ce qui signifie qu'il n'y a pas de déflexion spatiale : $\varphi$ reste constant, $\kappa$ diminue de $5$ à environ $3.4$, et la surface 3D est un hélicoïde immaculé. À mesure que $x_0$ grandit, les rayons avec un paramètre d'impact $|b| < 0.1$ traversent le cœur, tandis que les autres sont déviés. La tension visuelle augmente : l'angle du covecteur change et la surface se déforme. Loin du centre, la chorégraphie se stabilise à nouveau en un hélicoïde autour de la source en mouvement.

- **Rayons liés (Parties B et C) : Le rituel des points de rebroussement.**
  Pour $r \le 0.1$ et de petits $|x_0|$, les rayons sont piégés dans le cœur et oscillent radialement. Dans la vue $(x,y,\varphi)$, le covecteur s'inverse à chaque point de rebroussement, se manifestant par l'artefact de saut de $\pi$ décrit dans la Section 4.3. L'angle accumulé peut atteindre $6\pi$ en $t \in [0,1]$. La transition la plus frappante se produit autour de $|x_0| \approx 2$ : l'éventail passe de lié à non lié, et la structure visuelle change abruptement. De plus, dès que la source quitte le centre exact ($x_0 \neq 0$), les renversements discrets de $\pi$ sont remplacés par un enroulement continu, altérant la topologie de la surface du front d'onde.

- **Dépendance au rayon de la source $r$ : L'échelle macroscopique contre microscopique.**
  Les images de la Partie B devraient converger à mesure que $r \to 0$ (la limite de la source ponctuelle idéale). La transition géométrique critique se produit autour de $r \approx \sqrt{\varepsilon} = 0.1$. Lorsque $r \gg \sqrt{\varepsilon}$, le cercle des graines entoure le cœur singulier ; lorsque $r \ll \sqrt{\varepsilon}$, le cercle se trouve profondément à l'intérieur du cœur régularisé, sondant le gradient de potentiel abrupt. Dans la vue $(x,y,\varphi)$, les rayons liés radiaux ($x_0=0$) alternent entre deux hauteurs à leurs points de rebroussement, tandis que pour $x_0 \neq 0$ le covecteur s'enroule continûment, de sorte que l'image change abruptement dès que la source quitte le centre.

## 7. Remarque numérique

Les rayons sont intégrés avec un schéma symplectique à pas fixe du premier ordre (`n_steps=500` par défaut, pas de $0.002$). Près du cœur, la force est grande, donc l'énergie $p$ n'est conservée qu'approximativement. La plus grande dérive de l'énergie sur un rayon a été :

| Rayons | `n_steps` $=500$ | $2000$ | $4000$ | $8000$ |
|---|---|---|---|---|
| liés, $|\kappa|=1$ | $\approx 0.15$ | $\approx 0.04$ | n/a | $\approx 0.01$ |
| diffusion, $|\kappa|=5$, $r=0.1$ | $0.18$–$0.26$ | n/a | $0.02$–$0.03$ | n/a |

Ce sont des erreurs absolues. Pour $|\kappa|=5$, elles représentent environ $2$–$3\%$ de l'énergie à `n_steps=500` et environ $0.3$–$0.6\%$ à $4000$. Pour les rayons liés d'énergie autour de $-0.5$, elles peuvent atteindre des dizaines de pourcents à `n_steps=500`. Avant de tirer des conclusions, vérifiez qu'une image ne change pas lorsque `n_steps` est augmenté (par exemple à `4000`).

Dans la vue 3D, chaque rayon est dessiné avec au maximum environ 200 points, de sorte que le rendu reste léger même pour de grands `n_steps`.
