# Ray fans through a softened Coulomb singularity

**Goal.** Follow the wavefront set of a *point source* as it is moved along the $x$-axis through the centre of an attractive Coulomb-type potential, and see how the fan of rays depends on

- the **size $r$** of the source,
- the **energy** of the rays: **bound** rays ($p<0$, covector norm $1$) versus **scattering** rays ($p>0$, covector norm $5$),
- the **view** of the phase space: four 2D slices, or the cotangent bundle itself drawn as *base plane + circle fibre*.

---

## 1. The operator

We work in $\mathbb{R}^2$ with the symbol

$$
p(x,y,\xi,\eta)\;=\;\frac{\xi^{2}+\eta^{2}}{2m}\;-\;\frac{k}{\sqrt{x^{2}+y^{2}+\varepsilon}} .
$$

With the Kohn–Nirenberg quantization,

$$
(Pu)(x)=\frac{1}{(2\pi)^2}\iint e^{\,i x\cdot\xi}\,p(x,\xi)\,\hat u(\xi)\,d\xi
\quad\Longrightarrow\quad
P=-\frac{1}{2m}\Delta+V(x),\qquad V(x)=-\frac{k}{\sqrt{|x|^{2}+\varepsilon}}\qquad(\hbar=1),
$$

so $P$ is a Schrödinger operator with a **softened Coulomb potential**. Only the symbol `op.symbol` is used below.

| Parameter | Value | Meaning |
|---|---|---|
| $m$ | $1$ | mass |
| $k$ | $1$ | coupling (attractive) |
| $\varepsilon$ | $0.01$ | softening |

The singularity of the pure Coulomb symbol ($\varepsilon=0$, $V\sim -k/|x|$) at the origin is regularised by $\varepsilon$. The regularised potential has

- core radius $\sqrt{\varepsilon}=0.1$,
- depth $V(0)=-k/\sqrt{\varepsilon}=-10$,
- maximal force $F_{\max}=\dfrac{2k}{3\sqrt3\,\varepsilon}\approx 38.5$, reached at $\lvert x\rvert=\sqrt{\varepsilon/2}\approx 0.07$.

---

## 2. Rays = bicharacteristics

Singularities of a solution propagate along the Hamiltonian flow of $p$:

$$
\dot x=\frac{\partial p}{\partial \xi}=\frac{\xi}{m},\qquad
\dot y=\frac{\partial p}{\partial \eta}=\frac{\eta}{m},\qquad
\dot \xi=-\frac{\partial p}{\partial x}=-\frac{k\,x}{(x^{2}+y^{2}+\varepsilon)^{3/2}},\qquad
\dot \eta=-\frac{\partial p}{\partial y}=-\frac{k\,y}{(x^{2}+y^{2}+\varepsilon)^{3/2}} .
$$

Two quantities are conserved along each ray: the symbol itself, $p$ (the "energy" $E$ of the ray), and the angular momentum $L=x\eta-y\xi$ (the symbol is rotation invariant). The impact parameter of a ray with respect to the centre is $b=L/\lvert\kappa\rvert$, where $\lvert\kappa\rvert=\sqrt{\xi^2+\eta^2}$ is its covector norm.

> This is the **geometric (semiclassical) picture**: the animation shows what the theory of propagation of singularities predicts. It is not a simulation of a wave equation (no amplitudes, no interference).

---

## 3. The point source and the energy

A point source at $(x_0,0)$ contains *every* direction of the covector. It is modelled by $N=300$ seeds on a small circle of radius $r$ around the source, each carrying the outward covector of norm $\lvert\kappa\rvert=\sqrt{\xi_0^2+\eta_0^2}$:

$$
(x_j,y_j)=\bigl(x_0+r\cos\theta_j,\; r\sin\theta_j\bigr),\qquad
(\xi_j,\eta_j)=\lvert\kappa\rvert\bigl(\cos\theta_j,\;\sin\theta_j\bigr),\qquad
\theta_j=\frac{2\pi j}{N}.
$$

If `xi0=0, eta0=0`, the norm takes its default value $\lvert\kappa\rvert=1$. The energy of a seed is

$$
p_j=\frac{\lvert\kappa\rvert^{2}}{2m}-\frac{k}{\sqrt{(x_0+r\cos\theta_j)^2+r^{2}\sin^{2}\theta_j+\varepsilon}} ,
$$

and the angular momentum is $L=x_0\lvert\kappa\rvert\sin\theta_j$, i.e. the impact parameter is $b=x_0\sin\theta_j$.

Two consequences:

- **For a source at the centre** ($x_0=0$) all rays have the same energy $p_0(r)=\tfrac{\lvert\kappa\rvert^2}{2}-k/\sqrt{r^2+\varepsilon}$ and $L=0$: every ray is **radial** and passes through the core.
- **For a source at $x_0\neq 0$** the energy is approximately $p\approx\tfrac{\lvert\kappa\rvert^{2}}{2}-k/\sqrt{x_0^2+\varepsilon}$. Only rays with $\lvert b\rvert\lesssim\sqrt\varepsilon=0.1$ that head towards the centre cross the core.

### Two regimes

| Source at the centre | $r=1$ | $r=0.1$ | $r=0.01$ | $r=0.001$ |
|---|---|---|---|---|
| $p_0$ for $\lvert\kappa\rvert=1$ (`xi0=0`) | $-0.495$ | $-6.57$ | $-9.45$ | $-9.50$ |
| $p_0$ for $\lvert\kappa\rvert=5$ (`xi0=5`) | $11.5$ | $5.43$ | $2.55$ | $2.50$ |

- **Bound rays, $\lvert\kappa\rvert=1$.** At $x_0=0$ the energy is negative, so the rays cannot escape. They oscillate radially inside the turning radius $r_t$ defined by $V(r_t)=p_0$: $r_t\approx 2.0$ for $r=1$, and $r_t\approx 0.115,\;0.035,\;0.033$ for $r=0.1,\;0.01,\;0.001$, i.e. inside the core. As the source moves away from the centre the energy rises and **changes sign at $\lvert x_0\rvert\approx 2.0$**: beyond that the rays are unbound. The wide sweep $x_0\in(-5,5)$ therefore crosses the threshold $p=0$.
- **Scattering rays, $\lvert\kappa\rvert=5$.** The energy is positive everywhere along the sweep (from about $2.5$ to $5.4$ at $x_0=0$, and about $12.3$ far away). The rays escape to infinity and are only deflected by the attractive centre. Positive energy at the centre requires $\lvert\kappa\rvert>\sqrt{2mk/\sqrt\varepsilon}\approx 4.47$.

Compare $r$ with the core radius $\sqrt\varepsilon=0.1$:

- $r=1\gg\sqrt\varepsilon$: the seed circle surrounds the core;
- $r=0.1\approx\sqrt\varepsilon$: the circle has the size of the core;
- $r=0.01,\,0.001\ll\sqrt\varepsilon$: the circle lies deep inside the core, close to the ideal point source ($r\to0$).

## 4. Seeing the fibre bundle: the $(x,y,\varphi)$ view

### 4.1 Phase space as a bundle

The phase space $T^*\mathbb{R}^2=\mathbb{R}^2_{(x,y)}\times\mathbb{R}^2_{(\xi,\eta)}$ is a bundle over the **base** $\mathbb{R}^2_{(x,y)}$: above each point lies a **fibre**, the plane of covectors at that point. Using polar coordinates in the fibre,

$$
\xi=\kappa\cos\varphi,\qquad \eta=\kappa\sin\varphi,\qquad\text{so that}\qquad T^*\mathbb{R}^2=\{(x,y,\kappa,\varphi)\},
$$

where $\kappa=\lvert(\xi,\eta)\rvert$ is the covector norm and $\varphi$ the covector angle.

The wavefront set of a point source at $(x_0,0)$ is the **whole fibre** above $x_0$ (without its origin). The seeds of Section 3 are the circle $\kappa=\lvert\kappa\rvert$ of this fibre, with $\varphi=\theta_j$, displaced by the radius $r$.

### 4.2 The 3D view

With `projection='3d'`, `coords3d=('x','y','phi')` and `color_by='kappa'` the plot uses

- the **base** $(x,y)$ as the horizontal plane,
- the **covector angle** $\varphi$ as the vertical axis: it is the circle fibre,
- the **covector norm** $\kappa$ as the colour (with a colour bar).

All four coordinates of $T^*\mathbb{R}^2$ are therefore visible at once. At $t=0$ the green initial points form a (nearly) vertical line of height $2\pi$ above the source, namely the circle fibre cut open and unrolled to $\varphi\in[-\pi,\pi)$ (its two ends are the same point of the fibre). The flow carries it into a surface, and the red squares reached at $t=1$ trace the image of the circle.

The angle $\varphi$ is **unwrapped along each ray** (no jump at $\pm\pi$), so the vertical axis is the universal cover of the circle: a ray that winds around the centre keeps climbing. Two useful viewing directions:

- the **top view** (`view=(90,-90)`) is the projection to the base, i.e. the position panel; folds of this projection are the caustics, where neighbouring rays cross;
- the **side view** (`view=(0,-90)`) shows how the covector turns along the front.

`ortho=True` gives an orthographic camera, and `rotate=2` turns the camera slowly during the animation.

### 4.3 How $\varphi$ and $\kappa$ evolve

In polar coordinates of the fibre, the flow gives

$$
\dot\varphi=\frac{\xi\dot\eta-\eta\dot\xi}{\kappa^{2}}=\frac{V'(\rho)}{\rho\,\kappa^{2}}\,L,\qquad \rho=\sqrt{x^2+y^2},
$$

so the covector turns **only where the angular momentum $L$ is not zero**. For a source at the centre ($L=0$) every ray keeps its angle $\varphi$ constant: the surface is then a **helicoid** (a spiral ramp whose axis is the vertical line above the source). Away from the centre the helicoid is bent by the deflection of the rays.

Note that $\varphi$ is not defined where $\kappa=0$. This happens at the turning points of **bound radial rays**: the covector reverses, $\varphi$ jumps by $\pi$ at each turning point, and the unwrapping accumulates these jumps, so the angle climbs in steps of $\pi$. This is an artefact of the coordinate, not of the flow.

## 6. What to look for

These expectations follow from the equations above; compare them with the animations.

- **Scattering (Parts A and C).** There is no trapping and no repeated passages: the fan keeps expanding and decelerates as it leaves the well. At $x_0=0$ all rays are radial, so there is no deflection: $\varphi$ stays constant, $\kappa$ decreases from $5$ to about $3.4$ over $t\in[0,1]$, and the 3D surface is a helicoid. As $x_0$ grows, rays with $\lvert b\rvert=\lvert x_0\sin\theta\rvert<0.1$ that head inwards cross the core, and the others are deflected without entering it. For $x_0=1$, about 18 of the 300 rays have $\lvert b\rvert<0.1$; after $t=1$ the covector angle has changed by up to about $0.33$ rad ($19^\circ$), and $\kappa$ ranges from about $4.8$ to $6.5$. Far from the centre the surface stays close to a helicoid around the source.
- **Bound rays (Parts B and C).** For $r\le0.1$ and $\lvert x_0\rvert$ small, the rays remain trapped in the core and oscillate through the centre. In the $(x,y,\varphi)$ view the covector reverses at each turning point, with the artefact described in Section 4.3: $\varphi$ jumps by $\pi$ there, and in my test (source at the centre, $r=0.1$) the accumulated angle reached up to about $6\pi$ within $t\in[0,1]$. Around $\lvert x_0\rvert\approx2$ the fan switches from bound to unbound, so the pictures should change qualitatively there.
- **Dependence on $r$.** The pictures of Part B should converge as $r\to0$ (the ideal point source); the interesting transition is around $r\approx\sqrt\varepsilon$.

## 7. Numerical remark

The rays are integrated with a first-order fixed-step symplectic scheme (`n_steps=500` by default, step $0.002$). Near the core the force is large, so the energy $p$ is only conserved approximately. The largest energy drift over a ray was

| Rays | `n_steps` $=500$ | $2000$ | $4000$ | $8000$ |
|---|---|---|---|---|
| bound, $\lvert\kappa\rvert=1$ | $\approx0.15$ | $\approx0.04$ | n/a | $\approx0.01$ |
| scattering, $\lvert\kappa\rvert=5$, $r=0.1$ | $0.18$–$0.26$ | n/a | $0.02$–$0.03$ | n/a |

These are absolute errors. For $\lvert\kappa\rvert=5$ they are about $2$–$3\%$ of the energy at `n_steps=500` and about $0.3$–$0.6\%$ at $4000$. For bound rays of energy around $-0.5$ they can be tens of percent at `n_steps=500`. Before drawing conclusions, check that a picture does not change when `n_steps` is increased (for example to `4000`).

In the 3D view each ray is drawn with at most about 200 points, so the rendering stays light even for large `n_steps`.

````python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import HTML

# Assuming psiop is on the python path
from psiop import *

# Spatial and frequency variables
x, y, xi, eta = sp.symbols('x y xi eta', real=True)

A = 1
B = 1
m = 1
k = 1
alpha = 1
beta = 1

eps = 0.01

symbol_expr = (xi**2 + eta**2)/(2*m) - k/sp.sqrt(x**2 + y**2 + eps)

op = PseudoDifferentialOperator(
    expr=symbol_expr,
    vars_x=[x, y],
    mode='symbol',
    quantization='kohn-nirenberg',
    apply_backend='peetre',
)

print('Operator symbol:')
sp.pprint(op.symbol)
print(f'\ndimension           : {op.dim}')
print(f'quantization        : {op.quantization}')
print(f'apply backend       : {op.apply_backend}')
print(f'spatially dependent : {op._is_spatial_dependent()}')
````

````python
def ray_diagnostics(op, x0=0.0, r=0.1, kappa=1.0, theta=0.0, tspan=(0, 1), n_steps=8000, plot=True):
    """Follow ONE ray of the point-source fan and report how kappa and phi evolve."""
    z0 = (x0 + r*np.cos(theta), r*np.sin(theta), kappa*np.cos(theta), kappa*np.sin(theta))
    tr = op.bicharacteristic_flow(z0, tspan, dim=2, method='symplectic', n_steps=n_steps)
    t, xi_, eta_ = np.asarray(tr['t']), np.asarray(tr['xi']), np.asarray(tr['eta'])
    kap = np.hypot(xi_, eta_)
    phi = np.unwrap(np.arctan2(eta_, xi_))
    steps = np.diff(phi)
    jumps = np.where(np.abs(steps) > 1.0)[0]                 # steps of about +-pi
    print(f"n_steps={n_steps}: kappa_min={kap.min():.4f}, {len(jumps)} jump(s) of phi, "
          f"sizes/pi = {np.round(steps[jumps]/np.pi, 2).tolist()}, "
          f"phi_final - phi_0 = {(phi[-1]-phi[0])/np.pi:.2f} pi")
    if plot:
        fig, ax = plt.subplots(1, 2, figsize=(10, 3.2))
        ax[0].plot(t, kap);       ax[0].set_xlabel('t'); ax[0].set_ylabel('kappa')
        ax[1].plot(t, phi/np.pi); ax[1].set_xlabel('t'); ax[1].set_ylabel('phi / pi')
        plt.tight_layout(); plt.show()

ray_diagnostics(op, kappa=1.0, theta=0.0)   # bound, source at the centre
ray_diagnostics(op, kappa=1.0, theta=1.0)   # same, another direction
ray_diagnostics(op, kappa=5.0)              # scattering, source at the centre
````
