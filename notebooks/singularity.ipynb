# Microlocal Dynamics of a Smoothed Central Potential

## Mathematical Formulation

We consider the propagation of singularities for a pseudo-differential operator whose principal symbol corresponds to the classical Hamiltonian of a particle moving in a two-dimensional smoothed central potential (a regularized Kepler or Coulomb system). 

The symbol $p(x, y, \xi, \eta)$ on the phase space $T^*\mathbb{R}^2$ is defined as:

$$
p(x, y, \xi, \eta) = \underbrace{\frac{\xi^2 + \eta^2}{2m}}_{\text{Kinetic Energy}} - \underbrace{\frac{k}{\sqrt{x^2 + y^2 + \epsilon}}}_{\text{Regularized Potential}}
$$

where:
- $(x, y) \in \mathbb{R}^2$ are the spatial coordinates.
- $(\xi, \eta) \in \mathbb{R}^2$ are the conjugate frequency (momentum) variables.
- $m > 0$ is the mass of the particle.
- $k > 0$ is the coupling constant of the attractive potential.
- $\epsilon > 0$ is a small regularization parameter that resolves the singularity at the origin, ensuring the symbol remains smooth and the Hamiltonian flow remains well-defined everywhere.


```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import HTML

# Assuming psiop is on the python path
from psiop import *

# Spatial and frequency variables
x, y, xi, eta = sp.symbols('x y xi eta', real=True)

A = 1
B = 1
m = 1
k = 1
alpha = 1
beta = 1

eps = 0.01

symbol_expr = (xi**2 + eta**2)/(2*m) - k/sp.sqrt(x**2 + y**2 + eps)

op = PseudoDifferentialOperator(
    expr=symbol_expr,
    vars_x=[x, y],
    mode='symbol',
    quantization='kohn-nirenberg',
    apply_backend='peetre',
)

print('Operator symbol:')
sp.pprint(op.symbol)
print(f'\ndimension           : {op.dim}')
print(f'quantization        : {op.quantization}')
print(f'apply backend       : {op.apply_backend}')
print(f'spatially dependent : {op._is_spatial_dependent()}')
```
