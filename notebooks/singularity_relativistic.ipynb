# Ray fans through a softened Coulomb singularity, relativistic kinematics

**Goal.** Repeat the study of ray fans emitted by a *point source of fixed energy* near an attractive Coulomb-type potential, but with the **relativistic** kinetic energy $\sqrt{\xi^2+\eta^2+m^2}$ instead of $(\xi^2+\eta^2)/2m$. The symbol is now that of a genuine **pseudodifferential operator** (non-local, of order 1). What changes:

- a **speed limit**: the rays travel at speed $\kappa/\sqrt{\kappa^2+m^2}<1$;
- much **stronger scattering** at comparable energies, with deflection angles above $180^\circ$ (the rays loop around the centre);
- **fall to the centre**: a bound ray with angular momentum $L<k$ plunges into the core, a ray with $L>k$ never does;
- for $m\to\infty$ the picture must **reduce to the Schrödinger case** (a check).

| Part | Content |
|---|---|
| A | scattering fans, four 2D slices of the phase space |
| B | bound fans: fall to the centre |
| C | the cotangent bundle itself: base plane + circle fibre, in 3D |
| D | the non-relativistic limit $m\to\infty$ |
| E | diagnostic of a single ray |

## 1. The operator

We work in $\mathbb{R}^2$ with $\hbar=c=1$ and the symbol

$$
p(x,y,\xi,\eta)\;=\;\sqrt{\xi^{2}+\eta^{2}+m^{2}}\;-\;\frac{k}{\sqrt{x^{2}+y^{2}+\varepsilon}} .
$$

With the Kohn–Nirenberg quantization,

$$
(Pu)(x)=\frac{1}{(2\pi)^2}\iint e^{\,i x\cdot\xi}\,p(x,\xi)\,\hat u(\xi)\,d\xi
\quad\Longrightarrow\quad
P=\sqrt{-\Delta+m^{2}}+V(x),\qquad V(x)=-\frac{k}{\sqrt{|x|^{2}+\varepsilon}},
$$

the "square-root Klein–Gordon" (Salpeter) operator with a **softened Coulomb potential**. Only the symbol `op.symbol` is used below.

| Parameter | Value | Meaning |
|---|---|---|
| $m$ | $1$ | mass (rest energy $m$) |
| $k$ | $1$ | coupling (attractive) |
| $\varepsilon$ | $0.01$ | softening |

The regularised potential has core radius $\sqrt\varepsilon=0.1$ and depth $V(0)=-k/\sqrt\varepsilon=-10$.

## 2. Rays = bicharacteristics

Singularities of a solution propagate along the Hamiltonian flow of $p$:

$$
\dot x=\frac{\xi}{\sqrt{\kappa^{2}+m^{2}}},\qquad
\dot y=\frac{\eta}{\sqrt{\kappa^{2}+m^{2}}},\qquad
\dot \xi=-\frac{k\,x}{(x^{2}+y^{2}+\varepsilon)^{3/2}},\qquad
\dot \eta=-\frac{k\,y}{(x^{2}+y^{2}+\varepsilon)^{3/2}},\qquad \kappa=\sqrt{\xi^2+\eta^2}.
$$

The speed $\lvert\dot{\mathbf x}\rvert=\kappa/\sqrt{\kappa^2+m^2}$ is **always below $1$**. Two quantities are conserved along each ray: the energy $p=E$ and the angular momentum $L=x\eta-y\xi$ (the symbol is rotation invariant).

> **Two pictures.** In the classical sense of Hörmander the principal symbol of $P$ is $\lvert\xi\rvert$ (order 1; $V$ is of order 0), so the rays would be straight lines at speed 1 whatever $V$ is, and $V$ would only affect the amplitudes. The bent rays of this notebook are the **semiclassical picture**, in which $V$ is of the same order as the kinetic term. In both cases the animations show what the theory of propagation of singularities predicts; they are not simulations of a wave equation (no amplitudes, no interference).

## 3. A point source on an energy shell

A point source at $(x_0,0)$ contains *every* direction of the covector. It is modelled by $N$ seeds on a small circle of radius $r$ around the source, each carrying an outward covector **whose norm is chosen so that $p=E$**:

$$
(x_j,y_j)=\bigl(x_0+r\cos\theta_j,\; r\sin\theta_j\bigr),\qquad
(\xi_j,\eta_j)=\kappa_j\bigl(\cos\theta_j,\;\sin\theta_j\bigr),\qquad
\kappa_j=\sqrt{\bigl(E-V(x_j,y_j)\bigr)^{2}-m^{2}},\qquad
\theta_j=\frac{2\pi j}{N}.
$$

In `animate_wavefront_set` this is the option `energy=E`. The energy now includes the **rest energy**: at a point $x$ the lowest possible value of $p$ is $m+V(x)$, so a seed exists only where $E\ge m+V(x)$ (the other seeds are dropped). In particular

- **scattering** needs $E>m$ (the rays can reach infinity);
- **bound motion** needs $E<m$: the rays stay in the region $m+V(\rho)\le E$.

## 4. Choice of the parameters

### Scattering energies

For a plane wave of rays arriving from far away with impact parameter $b$, the deflection angle $\Theta(b)$ (computed numerically, continued across $180^\circ$) has a maximum $\Theta_{\max}$, where the rays pile up in a fold caustic (a rainbow):

| $E$ | $1.2$ | $1.5$ | $2$ | $3$ | $6$ | $12$ |
|---|---|---|---|---|---|---|
| $\Theta_{\max}$ (degrees) | $255$ | $208$ | $165$ | $121$ | $71$ | $40$ |
| at $b\approx$ | $1.38$ | $0.93$ | $0.65$ | $0.43$ | $0.25$ | $0.18$ |

For comparison, the Schrödinger case at $E=2$ peaks at $56^\circ$. Two energies are used:

- $E=3$: a clean rainbow of $121^\circ$ at $b\approx0.43$, and fast rays (speed $\ge0.95$);
- $E=1.5$: deflections above $180^\circ$ (for example $204^\circ$ at $b=1$): the rays loop around the centre.

For a source at $x_0$ the impact parameter of a ray is $b=x_0\sin\theta$, so $x_0\gtrsim0.43$ covers the rainbow of $E=3$.

### Bound energy $E=0.8$ and the fall to the centre

For $E=0.8<m$ the allowed region is $\rho\le5.0$ (where $m+V(\rho)=E$). For the relativistic Coulomb potential ($\varepsilon\to0$), a ray with angular momentum $L<k$ **reaches the centre**, while $L>k$ keeps it away (the "centrifugal barrier" $L/\rho$ wins over $k/\rho$). The softening stops the fall at the core. For a source at $x_0$ with initial norm $\kappa_0$, the angular momentum is $L=x_0\kappa_0\sin\theta$, so the fraction of falling rays is

$$
\frac{2}{\pi}\arcsin\frac{k}{x_0\kappa_0}\qquad (x_0\kappa_0>k).
$$

| source $x_0$ | $0.5$ | $1$ | $2$ | $3$ |
|---|---|---|---|---|
| $\kappa_0$ | $2.57$ | $1.49$ | $0.83$ | $0.53$ |
| rays with $L<k$ (of 300), measured | $170$ | $140$ | $124$ | $130$ |
| prediction | $170$ | $140$ | $124$ | $129$ |
| of these, within $0.05$ of the centre in $T=30$ | $74$ | $62$ | $58$ | $58$ |
| rays with $L>k$ within $0.05$ of the centre | $0$ of $130$ | $0$ of $160$ | $0$ of $176$ | $0$ of $170$ |

$T=30$ is about one revolution of the outer orbits.

### Source radius

Median distance between the final positions of the same ray for a source of radius $r$ and for the reference $r=0.001$:

| $r$ | $0.5$ | $0.1$ | $0.02$ | $0.01$ |
|---|---|---|---|---|
| A: $E=3$, $x_0=1$, $T=6$ | $0.86$ | $0.18$ | $0.036$ | $0.017$ |
| B: $E=0.8$, $x_0=1$, $T=30$ | $1.27$ | $0.24$ | $0.046$ | $0.022$ |

The error is of order $2r$: $r=0.02$ is a point source for the purposes of the pictures, $r=0.5$ is a ring-shaped source.

### Accuracy and cost

The animations use the second-order velocity-Verlet scheme (the symbol is separable). Largest energy drift $\max\lvert p-E\rvert$ over the rays and over the sweep: $2\cdot10^{-5}$ (A1, `n_steps=12000`), $3\cdot10^{-5}$ (A3, $16000$ steps), $8\cdot10^{-5}$ (B, $30000$ steps); the diagnostic of Part E, which uses symplectic Euler, drifts by about $10^{-2}$. Each of the animations below (24 frames, 300 seeds) builds in $3$–$6$ s and needs about $0.5$ GB.

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import HTML

# Assuming psiop is on the python path
from psiop import *
from caustics import * 

# Spatial and frequency variables
x, y, xi, eta = sp.symbols('x y xi eta', real=True)

m = 1
k = 1
eps = 0.01

symbol_expr = sp.sqrt(xi**2 + eta**2 + m**2) - k/sp.sqrt(x**2 + y**2 + eps)

op = PseudoDifferentialOperator(
    expr=symbol_expr,
    vars_x=[x, y],
    mode='symbol',
    quantization='kohn-nirenberg',
    apply_backend='peetre',
)

print('Operator symbol:')
sp.pprint(op.symbol)
print(f'\ndimension           : {op.dim}')
print(f'quantization        : {op.quantization}')
print(f'apply backend       : {op.apply_backend}')
print(f'spatially dependent : {op._is_spatial_dependent()}')

# Energies (rest energy m = 1: scattering for E > m, bound motion for E < m)
E_scat = 3.0      # rainbow of 121 degrees
E_loop = 1.5      # deflections above 180 degrees
E_bound = 0.8     # bound motion, allowed region rho <= 5
print(f'\nrest energy m = {m}: E_scat = {E_scat}, E_loop = {E_loop}, E_bound = {E_bound}')

plt.rcParams['animation.embed_limit'] = 2**128
```

- **At $x_0=0$** the angular momentum vanishes: every ray is radial and the front is a perfect circle, with no distortion at all.

## A5. The wedge edges are caustics

A rainbow caustic is a **fold** of the map from the initial direction $\theta$ of a ray to its final direction $\alpha$: at the edge $\alpha'(\theta_c)=0$ and $\alpha''(\theta_c)\neq0$, and the rays pile up with a density $\rho\propto s^{-1/2}$, where $s$ is the angular distance to the edge. The function below computes $\alpha(\theta)$ for a fan and locates its critical points; the critical rays (red) are the edges of the wedge, and their direction is the rainbow angle.

The measured turn of the edge direction from the axis tends to the rainbow angle of Section 4 as the source moves away (the small remaining gap is the part of the deflection missed by a source at finite distance):

| $x_0$ | $5$ | $20$ | plane wave |
|---|---|---|---|
| turn of the edge | $\approx116^\circ$ | $\approx120^\circ$ | $121^\circ$ |

The values at $x_0=5$ come from the cell above (symplectic Euler: $115.8^\circ$ and $116.2^\circ$); a dense computation with $20000$ rays and a second-order integrator gives $116.5^\circ$, and $120.2^\circ$ at $x_0=20$.

With $30000$ rays at $x_0=5$ the density behaves as $\rho\sqrt s\approx0.07$ over three decades of $s$ ($0.002$ to $0.06$ rad), which is the signature of a fold.

The caustic itself is found by the caustic detector of the package, which looks for the zeros of $\det J$ with $J=\partial(x,y)/\partial(\theta,t)$, the Jacobian of the ray map (computed here from neighbouring rays). For each ray it gives the crossing time, the position and the Maslov index:

- every ray whose initial direction lies **between the two critical angles crosses the caustic exactly once** (Maslov index $1$), and no other ray does;
- the crossings occur **close to the core** (median distance to the centre $0.23$ for the Schrödinger fan at $x_0=5$, $0.11$ for the relativistic one, to be compared with $\sqrt\varepsilon=0.1$) and move out along the critical rays as $\theta\to\theta_c$; the wedge edges are the asymptotes of the caustic.

```python
def caustic_edges(op, energy, x0, r=0.001, half_window=0.25, N=121, T=12.0, n_steps=8000, plot=True, detect=True):
    '''Wedge edges of the fan of a point source: folds of the final direction alpha(theta) (rainbow angle),
    and caustic events det J = 0 found by RayCausticDetector (position, time, Maslov index).'''
    p = sp.lambdify((x, y, xi, eta), op.symbol, 'numpy')
    c0 = np.pi if x0 > 0 else 0.0                                   # rays heading towards the centre
    th = c0 + np.linspace(-half_window, half_window, N)
    X, Y = x0 + r*np.cos(th), r*np.sin(th)
    lo, hi = np.zeros(N), np.ones(N)                                # covector norms: bisection on p = energy
    while np.any(p(X, Y, hi*np.cos(th), hi*np.sin(th)) < energy):
        hi = np.where(p(X, Y, hi*np.cos(th), hi*np.sin(th)) < energy, 2*hi, hi)
    for _ in range(80):
        mid = 0.5*(lo + hi); neg = p(X, Y, mid*np.cos(th), mid*np.sin(th)) < energy
        lo, hi = np.where(neg, mid, lo), np.where(neg, hi, mid)
    kap = 0.5*(lo + hi)
    rays = []
    for j in range(N):
        z0 = (X[j], Y[j], kap[j]*np.cos(th[j]), kap[j]*np.sin(th[j]))
        rays.append(bicharacteristic_flow(op.symbol, z0, (0, T), dim=2, method='symplectic', n_steps=n_steps))
    alpha = np.array([np.arctan2(tr['eta'][-1], tr['xi'][-1]) for tr in rays])      # final direction
    alpha = c0 + np.unwrap(np.angle(np.exp(1j*(alpha - c0))))                       # continuous in theta
    da = np.gradient(alpha, th)                                                     # d alpha / d theta
    crit = np.where(np.sign(da[:-1])*np.sign(da[1:]) < 0)[0]
    print(f"x0={x0}: {len(crit)} critical point(s) of alpha(theta); the wedge edges are folds if alpha'' != 0")
    found = []
    for i in crit:
        sl = slice(max(i-5, 0), min(i+6, N)); cf = np.polyfit(th[sl] - th[i], alpha[sl], 2)
        thc = th[i] - cf[1]/(2*cf[0]); ac = np.polyval(cf, thc - th[i])
        found.append((thc, ac)); off = np.degrees(abs(ac - np.pi))
        print(f"   theta_c={np.degrees(thc):8.3f} deg, alpha_c={np.degrees(ac):8.3f} deg  "
              f"(turn of the direction from the axis: {off:6.2f} deg), alpha''={2*cf[0]:9.2f}")
    events = []
    if detect:
        # Jacobian of the ray map (theta, t) -> (x, y), by finite differences between neighbouring rays.
        # It replaces the J11..J22 returned by bicharacteristic_flow, which obey dJ/dt = H_px J only
        # (identically J = 1 when p is separable, so no caustic would ever be detected).
        Xs = np.array([tr['x'] for tr in rays]); Ys = np.array([tr['y'] for tr in rays]); t = rays[0]['t']
        dXdth, dYdth = np.gradient(Xs, th, axis=0), np.gradient(Ys, th, axis=0)
        dXdt, dYdt = np.gradient(Xs, t, axis=1), np.gradient(Ys, t, axis=1)
        for j, tr in enumerate(rays):
            tr['J11'], tr['J21'], tr['J12'], tr['J22'] = dXdth[j], dYdth[j], dXdt[j], dYdt[j]
        det = RayCausticDetector(rays, 2, det_threshold=0.05)
        events = det.detect()
        ids = np.array([e.ray_idx for e in events], dtype=int)
        if len(events):
            tt = np.array([e.time for e in events]); pos = np.array([e.position for e in events]); dist = np.hypot(pos[:, 0], pos[:, 1])
            mas = sorted({det.maslov_index(int(i)) for i in set(ids)})
            print(f"   caustic events (det J = 0): {len(events)} on {len(set(ids))} of {N} rays, "
                  f"theta in [{np.degrees(th[ids].min()):.2f}, {np.degrees(th[ids].max()):.2f}] deg, Maslov index {mas}")
            print(f"   time: min {tt.min():.2f}, median {np.median(tt):.2f}, max {tt.max():.2f};  "
                  f"distance to the centre: min {dist.min():.2f}, median {np.median(dist):.2f}, max {dist.max():.2f}")
        else:
            print('   no caustic event')
    if plot:
        fig, ax = plt.subplots(1, 2, figsize=(11, 4.2))
        for tr in rays:
            ax[0].plot(tr['x'], tr['y'], lw=0.4, color='0.6')
        for i in crit:
            ax[0].plot(rays[i]['x'], rays[i]['y'], lw=1.5, color='crimson')
        if len(events):
            ax[0].plot(pos[:, 0], pos[:, 1], 'b.', ms=3)
        ax[0].plot([x0], [0], 'g.', ms=10); ax[0].plot([0], [0], 'k+'); ax[0].set_aspect('equal')
        ax[0].set_title('rays (grey), critical rays (red), caustic events (blue)')
        ax[1].plot(np.degrees(th), np.degrees(alpha - np.pi)); ax[1].set_xlabel('initial direction theta (deg)')
        ax[1].set_ylabel('final direction - 180 deg'); ax[1].set_title('alpha(theta): the edges are its extrema')
        for thc, ac in found:
            ax[1].plot(np.degrees(thc), np.degrees(ac - np.pi), 'ro')
        plt.tight_layout(); plt.show()
    return found
```

# Part B: bound motion, $E=0.8$

Flow time $T=30$ (about one revolution of the outer orbits), sources in the classically allowed region $\rho\le5$.

**What to look for.** The outer circle of radius $5$ is the turning radius. The rays with $L>k$ never come close to the centre: they form precessing rosettes (relativistic orbits are not closed ellipses). The rays with $L<k$ plunge into the core and come out on the other side, which shows as large excursions in the $(\xi,\eta)$ panel. The proportion of falling rays is $(2/\pi)\arcsin(k/x_0\kappa_0)$: $57\%$, $47\%$, $41\%$, $43\%$ for $x_0=0.5,1,2,3$.

# Part C: the cotangent bundle, base plane + circle fibre

The phase space $T^*\mathbb{R}^2=\mathbb{R}^2_{(x,y)}\times\mathbb{R}^2_{(\xi,\eta)}$ is a bundle over the base plane $(x,y)$; in polar coordinates of the fibre, $\xi=\kappa\cos\varphi$, $\eta=\kappa\sin\varphi$. With `coords3d=('x','y','phi')` and `color_by='kappa'` the plot shows the base $(x,y)$ as the horizontal plane, the covector angle $\varphi$ as the vertical axis (the circle fibre), and the covector norm $\kappa$ as the colour. At $t=0$ the green points form a (nearly) vertical line of height $2\pi$ above the source (the whole fibre, cut open at $\varphi=\pm\pi$); the flow carries it into a surface. The angle is made continuous along each ray (only the $2\pi$ wraps are removed).

The covector turns only where the angular momentum is not zero:

$$
\dot\varphi=\frac{\xi\dot\eta-\eta\dot\xi}{\kappa^{2}}=\frac{V'(\rho)}{\rho\,\kappa^{2}}\,L,\qquad \rho=\sqrt{x^2+y^2}.
$$

For a source at the centre ($L=0$) $\varphi$ stays constant on every ray and the scattering surface is a **helicoid**. At a turning point of a radial ray the covector vanishes and reverses, so $\varphi$ jumps by $\pi$ there (an artefact of the angle coordinate).

# Part D: the non-relativistic limit

Writing $E=m+E_{\rm nr}$ and $\kappa\ll m$,

$$
\sqrt{\kappa^2+m^2}\simeq m+\frac{\kappa^{2}}{2m},
$$

so the symbol is $m$ plus the Schrödinger symbol of mass $m$, and a constant does not change the flow: for $m\to\infty$ the relativistic fan of energy $m+E_{\rm nr}$ must approach the Schrödinger fan of energy $E_{\rm nr}$ (the first notebook). The function below compares the two for $E_{\rm nr}=2$, a source at $x_0=1$ and flow time $T=3$, and reports the largest distance between the final positions of corresponding rays.

```python
def nonrel_limit_check(masses=(1, 3, 10, 30), E_nr=2.0, x0=1.0, r=0.02, T=3.0, N=24, n_steps=2000):
    '''Relativistic fan with energy m + E_nr versus Schrodinger fan of mass m and energy E_nr.'''
    th = np.linspace(0, 2*np.pi, N, endpoint=False)
    X, Y = x0 + r*np.cos(th), r*np.sin(th)
    V = -k/np.sqrt(X**2 + Y**2 + eps)
    for mm in masses:
        p_rel = sp.sqrt(xi**2 + eta**2 + mm**2) - k/sp.sqrt(x**2 + y**2 + eps)
        p_nr = (xi**2 + eta**2)/(2*mm) - k/sp.sqrt(x**2 + y**2 + eps)
        kap_rel = np.sqrt((mm + E_nr - V)**2 - mm**2)
        kap_nr = np.sqrt(2*mm*(E_nr - V))
        dist, size = [], []
        for j in range(N):
            c, s_ = np.cos(th[j]), np.sin(th[j])
            fr = bicharacteristic_flow(p_rel, (X[j], Y[j], kap_rel[j]*c, kap_rel[j]*s_), (0, T), dim=2, method='symplectic', n_steps=n_steps)
            fn = bicharacteristic_flow(p_nr, (X[j], Y[j], kap_nr[j]*c, kap_nr[j]*s_), (0, T), dim=2, method='symplectic', n_steps=n_steps)
            dist.append(np.hypot(fr['x'][-1] - fn['x'][-1], fr['y'][-1] - fn['y'][-1]))
            size.append(np.hypot(fn['x'][-1] - x0, fn['y'][-1]))
        print(f'm={mm:4d}: max distance between the two fans = {max(dist):7.3f}  '
              f'({100*max(dist)/np.mean(size):5.1f}% of the distance travelled, {np.mean(size):.2f})')

nonrel_limit_check()
```

### Expected output

| $m$ | max distance between the two fans | distance travelled by the Schrödinger rays | ratio |
|---|---|---|---|
| $1$ | $4.895$ | $6.60$ | $74.1\%$ |
| $3$ | $2.111$ | $3.96$ | $53.4\%$ |
| $10$ | $0.574$ | $2.27$ | $25.3\%$ |
| $30$ | $0.159$ | $1.37$ | $11.6\%$ |

The relative difference decreases steadily (roughly like $1/m$): the two notebooks are consistent. Taking the Schrödinger rays of mass $m$ with $m=1$ is not a good approximation of the relativistic ones: at the energies of Part A the kinetic energy is comparable to the rest energy.

# Part E: diagnostic of a single ray

The function follows **one ray** of the point-source fan on the energy shell $p=E$ and prints

- the initial covector norm $\kappa_0$ and the angular momentum $L$ (compare with $k=1$);
- the range of the speed $\kappa/\sqrt{\kappa^2+m^2}$ (always below $1$);
- the range of the distance $\rho$ to the centre;
- the energy drift $\max\lvert p-E\rvert$;
- the total rotation of the covector, in degrees and in units of $\pi$, which is the deflection angle of the ray (at a turning point, $\kappa=0$, the covector reverses and the rotation is $180^\circ$).

The ray is integrated with `bicharacteristic_flow` (symplectic Euler), so its energy drift is larger than in the animations.

```python
def ray_diagnostics(op, energy, x0=0.0, r=0.02, theta=0.0, tspan=(0, 1), n_steps=8000, plot=True):
    '''Follow ONE ray of the point-source fan on the energy shell p = energy.'''
    p = sp.lambdify((x, y, xi, eta), op.symbol, 'numpy')
    X, Y = x0 + r*np.cos(theta), r*np.sin(theta)
    if p(X, Y, 0.0, 0.0) >= energy:
        raise ValueError('classically forbidden: p(x, y, 0, 0) >= energy at the seed')
    lo, hi = 0.0, 1.0                                         # covector norm: bisection on p = energy
    while p(X, Y, hi*np.cos(theta), hi*np.sin(theta)) < energy:
        hi *= 2
    for _ in range(80):
        mid = 0.5*(lo + hi)
        lo, hi = (mid, hi) if p(X, Y, mid*np.cos(theta), mid*np.sin(theta)) < energy else (lo, mid)
    kappa = 0.5*(lo + hi)
    z0 = (X, Y, kappa*np.cos(theta), kappa*np.sin(theta))
    tr = bicharacteristic_flow(op.symbol, z0, tspan, dim=2, method='symplectic', n_steps=n_steps)
    t, x_, y_, xi_, eta_ = (np.asarray(tr[c]) for c in ('t', 'x', 'y', 'xi', 'eta'))
    kap = np.hypot(xi_, eta_)
    speed = kap/np.sqrt(kap**2 + m**2)                        # |dx/dt| < 1
    rho = np.hypot(x_, y_)
    L = x_*eta_ - y_*xi_
    phi = np.unwrap(np.arctan2(eta_, xi_))
    drift = np.abs(p(x_, y_, xi_, eta_) - energy).max()
    print(f'kappa_0={kappa:.3f}, L={L[0]:.3f} (k={k}), speed in [{speed.min():.3f}, {speed.max():.4f}], '
          f'rho in [{rho.min():.3f}, {rho.max():.3f}], energy drift={drift:.1e}, '
          f'rotation of the covector={np.degrees(phi[-1] - phi[0]):.0f} deg ({(phi[-1] - phi[0])/np.pi:.2f} pi)')
    if plot:
        fig, ax = plt.subplots(1, 2, figsize=(10, 3.2))
        ax[0].plot(t, rho);   ax[0].set_xlabel('t'); ax[0].set_ylabel('distance to the centre')
        ax[1].plot(t, speed); ax[1].set_xlabel('t'); ax[1].set_ylabel('speed')
        plt.tight_layout(); plt.show()

ray_diagnostics(op, E_bound, x0=2.0, r=0.01, theta=0.0, tspan=(0, 30), n_steps=30000)            # bound, L = 0
ray_diagnostics(op, E_bound, x0=2.0, r=0.01, theta=np.pi/2, tspan=(0, 30), n_steps=30000)        # bound, L > k
ray_diagnostics(op, E_scat, x0=1.0, r=0.02, theta=np.pi, tspan=(0, 6), n_steps=12000)            # scattering, head-on
ray_diagnostics(op, E_scat, x0=1.0, r=0.02, theta=np.pi - np.arcsin(0.43), tspan=(0, 6), n_steps=12000)   # b = 0.43
```

| bound, $x_0=2$, $\theta=\pi/2$ | $0.830$ | $1.659$ | $[0.560,\ 0.639]$ | $[2.000,\ 2.454]$ | $5.5\cdot10^{-6}$ | $465^\circ$ ($2.58\,\pi$) |
