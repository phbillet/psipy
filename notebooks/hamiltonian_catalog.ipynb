## 1D Hamiltonian analysis

```python
NAME   = 'anharmonic_oscillator'
PARAMS = {'m': 1.0, 'alpha': 0.1}

# ── Catalog info ─────────────────────────────────────────────────────────
print_hamiltonian_info(NAME)

# ── 1. Instantiate the ΨDO ───────────────────────────────────────────────
op = get_operator(NAME, param_values=PARAMS, quantization='weyl')
print(f"[get_operator]  symbol  = {op.symbol}")
print(f"[get_operator]  dim     = {op.dim}")
print(f"[get_operator]  quant.  = {op.quantization}")

# ── 2. Full ΨDO analysis ─────────────────────────────────────────────────
x_grid  = np.linspace(-4, 4, 60)
xi_grid = np.linspace(-4, 4, 60)

analysis = analyze_hamiltonian(NAME, param_values=PARAMS,
                               x_grid=x_grid, xi_grid=xi_grid)

print("\n[analyze_hamiltonian]")
print(f"  Operator order       : {analysis['order']}")
print(f"  Is elliptic          : {analysis['is_elliptic']}")
print(f"  Is self-adjoint      : {analysis['is_self_adjoint']}")
print(f"  Principal symbol     : {analysis['principal_symbol']}")
print(f"  Formal adjoint       : {analysis['formal_adjoint']}")
print(f"  Symplectic flow      :")
for key, val in analysis['symplectic_flow'].items():
    print(f"    {key} = {val}")

# ── 3. Semiclassical trace ───────────────────────────────────────────────
# Symbolic trace  (closed-form integral over phase space)
trace_sym = trace_hamiltonian(NAME, param_values=PARAMS, numerical=False)
print(f"\n[trace_hamiltonian — symbolic]\n  Tr(H) = {trace_sym}")

# Numerical trace over a bounded region
trace_num = trace_hamiltonian(
    NAME, param_values=PARAMS,
    numerical=True,
    x_bounds=(-3, 3), xi_bounds=(-3, 3),
)
print(f"[trace_hamiltonian — numerical, x∈[-3,3], ξ∈[-3,3]]\n  Tr(H) ≈ {trace_num:.6f}")

# ── 4. Visualisations ────────────────────────────────────────────────────
interactive_hamiltonian(name=NAME, param_values=PARAMS)
```

## 2D Hamiltonian analysis

```python
NAME_HH   = 'henon_heiles'
NAME_FREE = 'isotropic_oscillator'       # used for composition
PARAMS_FREE = {'alpha': 1.0}

# ── Catalog info ─────────────────────────────────────────────────────────
print_hamiltonian_info(NAME_HH)

# ── 1. Instantiate the ΨDO ───────────────────────────────────────────────
op = get_operator(NAME_HH, param_values=None, quantization='weyl')
print(f"[get_operator]  symbol  = {op.symbol}")
print(f"[get_operator]  dim     = {op.dim}")

# ── 2. Full ΨDO analysis ─────────────────────────────────────────────────
grid_xy  = np.linspace(-2, 2, 40)
grid_mom = np.linspace(-3, 3, 40)

# For 2D: x_grid = (x_array, y_array), xi_grid = (xi_array, eta_array)
analysis = analyze_hamiltonian(
    NAME_HH, param_values=PARAMS_FREE,
    x_grid=(grid_xy, grid_xy),
    xi_grid=(grid_mom, grid_mom),
)

print("\n[analyze_hamiltonian]")
print(f"  Operator order       : {analysis['order']}")
print(f"  Is elliptic          : {analysis['is_elliptic']}")
print(f"  Is self-adjoint      : {analysis['is_self_adjoint']}")
print(f"  Principal symbol     : {analysis['principal_symbol']}")
print(f"  Formal adjoint       : {analysis['formal_adjoint']}")
print(f"  Symplectic flow      :")
for key, val in analysis['symplectic_flow'].items():
    print(f"    {key} = {val}")

# ── 3. Semiclassical trace (numerical) ───────────────────────────────────
# Symbolic integration is intractable for Hénon–Heiles (non-polynomial sqrt);
# we use the numerical path over a compact phase-space region.
trace_num = trace_hamiltonian(
    NAME_HH, param_values=PARAMS_FREE,
    numerical=True,
    x_bounds=((-2, 2), (-2, 2)), xi_bounds=((-2, 2), (-2, 2)),
)
print(f"\n[trace_hamiltonian — numerical, (x,y)∈[-2,2]², (ξ,η)∈[-2,2]²]")
print(f"  Tr(H_HH) ≈ {trace_num:.6f}")

# ── 4. Visualisations ────────────────────────────────────────────────────
interactive_hamiltonian(name=NAME_HH, param_values=PARAMS_FREE, density=10)
```
