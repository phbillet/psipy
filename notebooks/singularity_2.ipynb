# Ray fans through a softened Coulomb singularity — accuracy-oriented parameters

Goal. Follow the wavefront set of a point source as it is moved along the $x$-axis through the centre of an attractive Coulomb-type potential, and see how the fan of rays depends on

- the size $r$ of the source, chosen relative to the regularised core radius $a=\sqrt{\varepsilon}$,
- the energy of the rays: bound rays ($p<0$, explicit covector norm $\kappa=1$) versus scattering rays ($p>0$, explicit covector norm $\kappa=3.6$),
- the view of phase space: four 2D slices, or the cotangent bundle itself drawn as base plane + circle fibre.

---

## 1. The operator

We work in $\mathbb{R}^2$ with the symbol

$$
p(x,y,\xi,\eta)
=
\frac{\xi^{2}+\eta^{2}}{2m}
-
\frac{k}{\sqrt{x^{2}+y^{2}+\varepsilon}}.
$$

With the Kohn–Nirenberg quantization,

$$
(Pu)(x)
=
\frac{1}{(2\pi)^2}
\iint
e^{i x\cdot \xi}
p(x,\xi)\widehat{u}(\xi)
\,d\xi
\quad\Longrightarrow\quad
P
=
-\frac{1}{2m}\Delta + V(x),
$$

where

$$
V(x)
=
-\frac{k}{\sqrt{|x|^{2}+\varepsilon}}.
$$

Thus $P$ is a Schrödinger operator with a softened Coulomb potential. Only the symbol `op.symbol` is used below.

| Parameter | Value | Meaning |
|---:|---:|---|
| $m$ | $1$ | mass |
| $k$ | $1$ | attractive coupling |
| $\varepsilon$ | $0.04$ | softening, chosen for better numerical accuracy |

The regularised potential has

- core radius

$$
a=\sqrt{\varepsilon}=0.2,
$$

- depth

$$
V(0)=-\frac{k}{\sqrt{\varepsilon}}=-5,
$$

- maximal force

$$
F_{\max}
=
\frac{2k}{3\sqrt{3}\,\varepsilon}
\approx 9.62,
$$

reached at

$$
|x|=\sqrt{\frac{\varepsilon}{2}}\approx 0.141.
$$

This choice is much easier to integrate than the original $\varepsilon=0.01$, for which $F_{\max}\approx 38.5$, while preserving the same qualitative regularised Coulomb structure. If a more singular potential is required, set $\varepsilon=0.01$, scale the radii by $1/2$, use approximately $\kappa=4.8$ for scattering rays, and increase `n_steps`.

---

## 2. Rays = bicharacteristics

Singularities of a solution propagate along the Hamiltonian flow of $p$:

$$
\dot{x}
=
\frac{\partial p}{\partial \xi}
=
\frac{\xi}{m},
\qquad
\dot{y}
=
\frac{\partial p}{\partial \eta}
=
\frac{\eta}{m},
$$

and

$$
\dot{\xi}
=
-\frac{\partial p}{\partial x}
=
-\frac{kx}{(x^{2}+y^{2}+\varepsilon)^{3/2}},
\qquad
\dot{\eta}
=
-\frac{\partial p}{\partial y}
=
-\frac{ky}{(x^{2}+y^{2}+\varepsilon)^{3/2}}.
$$

Two quantities are conserved along each ray:

- the symbol itself,

$$
p=E,
$$

which is the ray energy;

- the angular momentum,

$$
L=x\eta-y\xi,
$$

because the symbol is rotation-invariant.

The covector norm is

$$
|\kappa|=\sqrt{\xi^{2}+\eta^{2}},
$$

and the impact parameter of a ray with respect to the centre is

$$
b=\frac{L}{|\kappa|}.
$$

> This is the geometric, semiclassical picture: the animation shows what propagation of singularities predicts. It is not a full wave simulation, and there are no amplitudes or interference terms.

---

## 3. The point source and the energy

A point source at $(x_0,0)$ contains every covector direction. It is modelled by $N=300$ seeds on a small circle of radius $r$ around the source. Each seed carries an outward covector of norm

$$
|\kappa|=\sqrt{\xi_0^2+\eta_0^2}.
$$

In the improved notebook this norm is chosen explicitly:

- bound rays: `xi0=1.0`, hence $|\kappa|=1$;
- scattering rays: `xi0=3.6`, hence $|\kappa|=3.6$.

The seeds are

$$
(x_j,y_j)
=
\bigl(
x_0+r\cos\theta_j,\;
r\sin\theta_j
\bigr),
$$

$$
(\xi_j,\eta_j)
=
|\kappa|
\bigl(
\cos\theta_j,\;
\sin\theta_j
\bigr),
$$

with

$$
\theta_j=\frac{2\pi j}{N}.
$$

The energy of a seed is

$$
p_j
=
\frac{|\kappa|^{2}}{2m}
-
\frac{k}{
\sqrt{
(x_0+r\cos\theta_j)^2
+
r^{2}\sin^{2}\theta_j
+
\varepsilon
}
}.
$$

The angular momentum is

$$
L_j
=
x_0|\kappa|\sin\theta_j,
$$

so the impact parameter is

$$
b_j=x_0\sin\theta_j.
$$

### Source at the centre

If $x_0=0$, every ray has $L=0$ and is radial. All seeds have the same energy

$$
p_0(r)
=
\frac{|\kappa|^2}{2}
-
\frac{k}{\sqrt{r^2+\varepsilon}}.
$$

### Source away from the centre

If $x_0\neq 0$, the energy is approximately

$$
p
\approx
\frac{|\kappa|^2}{2}
-
\frac{k}{\sqrt{x_0^2+\varepsilon}}.
$$

Only rays with impact parameter

$$
|b|\lesssim \sqrt{\varepsilon}=0.2
$$

and directed toward the centre cross the regularised core.

## 5. Two energy regimes

The following table gives the central energies $p_0(r)$ at $x_0=0$.

| radius $r$ | bound rays, $|\kappa|=1$ | scattering rays, $|\kappa|=3.6$ |
|---:|---:|---:|
| $0.60$ | $-1.081$ | $4.899$ |
| $0.20$ | $-3.036$ | $2.944$ |
| $0.06$ | $-4.289$ | $1.691$ |
| $0.02$ | $-4.475$ | $1.505$ |

### Bound rays

For bound rays we use

$$
|\kappa|=1.
$$

At $x_0=0$ the energy is negative, so the rays cannot escape. They oscillate between turning points. The outer turning radius $r_t$ is defined by

$$
V(r_t)=p_0,
$$

that is,

$$
-\frac{k}{\sqrt{r_t^2+\varepsilon}}=p_0.
$$

Since $p_0<0$, this gives

$$
r_t
=
\sqrt{
\left(\frac{k}{|p_0|}\right)^2
-
\varepsilon
}.
$$

For the four chosen radii and $|\kappa|=1$, the turning radii are approximately

$$
r_t
\approx
0.903,\quad
0.262,\quad
0.120,\quad
0.100.
$$

As the source moves away from the centre, the approximate energy is

$$
p(x_0)
\approx
\frac{1}{2}
-
\frac{1}{\sqrt{x_0^2+0.04}}.
$$

The bound-to-unbound transition occurs when $p(x_0)=0$, hence

$$
\frac{1}{2}
=
\frac{1}{\sqrt{x_0^2+0.04}},
$$

so

$$
x_0^2+0.04=4,
$$

and therefore

$$
|x_0|
=
\sqrt{3.96}
\approx 1.99.
$$

Thus the wide sweep $x_0\in(-4,4)$ shows the transition from bound to unbound rays.

### Scattering rays

For scattering rays we use

$$
|\kappa|=3.6.
$$

The energy is positive everywhere in the sweep. At the deepest point of the regularised potential, positive energy requires approximately

$$
|\kappa|
>
\sqrt{\frac{2k}{\sqrt{\varepsilon}}}
=
\sqrt{10}
\approx 3.162.
$$

The value $|\kappa|=3.6$ is safely above this threshold, while still producing more visible deflection than the original choice $|\kappa|=5$.

## 6. The cotangent-bundle view: the $(x,y,\varphi)$ picture

### 6.1 Phase space as a kinetic bundle

The phase space is

$$
T^*\mathbb{R}^2
=
\mathbb{R}^2_{(x,y)}
\times
\mathbb{R}^2_{(\xi,\eta)}.
$$

Above each spatial point $(x,y)$ lies the fibre of covectors $(\xi,\eta)$. Introducing polar coordinates in the fibre,

$$
\xi=\kappa\cos\varphi,
\qquad
\eta=\kappa\sin\varphi,
$$

separates the kinetic scale

$$
\kappa=\sqrt{\xi^2+\eta^2}
$$

from the directional phase $\varphi$.

The wavefront set of a point source at $(x_0,0)$ is the entire punctured fibre above $x_0$. The seeds form the circle

$$
\kappa=|\kappa_0|
$$

inside this fibre, with initial angles

$$
\varphi=\theta_j,
$$

displaced physically by the source radius $r$.

### 6.2 The 3D representation

With

```python
projection='3d'
coords3d=('x', 'y', 'phi')
color_by='kappa'
```

the plot represents:

- the spatial base $(x,y)$ as the horizontal plane;
- the covector angle $\varphi$ as the vertical axis;
- the covector norm $\kappa$ as colour.

At $t=0$, the seeds form a vertical line of height $2\pi$ above the source. This is the circle fibre cut open and unrolled to

$$
\varphi\in[-\pi,\pi).
$$

The Hamiltonian flow carries this line into a two-dimensional surface.

Two viewpoints are especially useful:

- the top view, which projects onto the base $(x,y)$ and reveals spatial caustics;
- the side view, which reveals momentum winding and covector rotation.

The option

```python
ortho=True
```

uses an orthographic camera, and

```python
rotate=2
```

rotates the camera slowly during the animation.

### 6.3 Evolution of $\varphi$ and $\kappa$

In fibre polar coordinates, the angle evolves according to

$$
\dot{\varphi}
=
\frac{\xi\dot{\eta}-\eta\dot{\xi}}{\kappa^2}
=
\frac{V'(\rho)}{\rho\,\kappa^2}L,
\qquad
\rho=\sqrt{x^2+y^2}.
$$

Therefore the covector direction turns only when

$$
L\neq 0.
$$

For a source exactly at the centre, $x_0=0$, all rays have

$$
L=0.
$$

Thus $\varphi$ is constant except at turning points of bound radial rays, where

$$
\kappa=0.
$$

At such points the polar coordinate $\varphi$ is undefined, and the reversal of the covector appears as a jump of size $\pi$. This is a coordinate artefact, not a physical rotation.

As soon as $x_0\neq 0$, angular momentum is generically nonzero, $\kappa$ does not vanish, and the discrete $\pi$-reversals are replaced by continuous winding.

## 12. What to look for

### Scattering rays

For scattering rays there is no trapping. The fan expands and decelerates as it climbs out of the potential well.

At $x_0=0$, all rays are radial and

$$
L=0.
$$

Therefore there is no angular deflection, and $\varphi$ remains constant. The covector norm decreases as the ray moves outward.

For the case

$$
r=0.20,
\qquad
|\kappa|=3.6,
$$

the central energy is

$$
p_0\approx 2.944.
$$

The asymptotic covector norm is therefore

$$
|\kappa|_{\infty}
=
\sqrt{2p_0}
\approx 2.43.
$$

As $x_0$ moves away from zero, rays with small impact parameter cross or graze the core, while the others are deflected. The covector angle changes, and the 3D surface warps. Far from the centre, the structure approaches a helicoid around the moving source.

### Bound rays

For bound rays with small $|x_0|$, the rays are trapped and oscillate between turning points. In the exact-centre case, $x_0=0$, the rays are radial and $\kappa$ vanishes at turning points. At those points the angle $\varphi$ jumps by approximately $\pi$.

The number of jumps depends on $r$, the flow time, and `n_steps`. The robust qualitative signature is:

$$
\kappa\to 0
\quad\text{and}\quad
\varphi\to\varphi+\pi.
$$

As soon as $x_0\neq 0$, angular momentum is generally nonzero. Then $\kappa$ no longer vanishes, and the discrete reversals are replaced by continuous winding.

The main global transition occurs near

$$
|x_0|\approx 1.99,
$$

where the fan changes from bound to unbound.

### Dependence on the source radius

The critical scale is the core radius

$$
a=\sqrt{\varepsilon}=0.2.
$$

The chosen radii illustrate the main geometric regimes:

- $r=0.60=3a$: the source circle surrounds the core;
- $r=0.20=a$: the source circle has the size of the core;
- $r=0.06=0.3a$: the source circle lies inside the core;
- $r=0.02=0.1a$: the source circle is close to the ideal point-source limit.

## 13. Numerical remark

The rays are integrated with a first-order fixed-step symplectic scheme. The improved notebook uses larger step counts than the original default `n_steps=500`:

| purpose | recommended `n_steps` |
|---|---:|
| quick exploration | $1000$--$2000$ |
| animations shown in the notebook | $4000$ |
| single-ray diagnostics and final checks | $8000$ |

With

$$
\varepsilon=0.04,
$$

the maximal force is approximately

$$
F_{\max}\approx 9.62,
$$

instead of

$$
F_{\max}\approx 38.5
$$

for $\varepsilon=0.01$. Therefore the same time step gives better energy behaviour.

Nevertheless, before interpreting a fine geometric feature, one should check that it does not disappear when `n_steps` is increased, for example from $4000$ to $8000$.

In the 3D view each ray is drawn with a limited number of graphical points, so the rendering remains light even when the ray integration itself uses a larger `n_steps`.

````python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import HTML

# Adjust this import if psiop exposes the names differently.
from psiop import PseudoDifferentialOperator, animate_wavefront_set, bicharacteristic_flow

x, y, xi, eta = sp.symbols('x y xi eta', real=True)

m = 1.0
k = 1.0

# Accuracy-oriented default: less stiff than eps = 0.01, but still a clear softened core.
eps = 0.04
core_radius = float(np.sqrt(eps))

symbol_expr = (xi**2 + eta**2)/(2*m) - k/sp.sqrt(x**2 + y**2 + eps)

op = PseudoDifferentialOperator(
    expr=symbol_expr,
    vars_x=[x, y],
    mode='symbol',
    quantization='kohn-nirenberg',
    apply_backend='peetre',
)

print('Operator symbol:')
sp.pprint(op.symbol)
print(f'dimension           : {op.dim}')
print(f'quantization        : {op.quantization}')
print(f'apply backend       : {op.apply_backend}')
print(f'spatially dependent : {op._is_spatial_dependent()}')
print(f'eps                 : {eps}')
print(f'core radius         : {core_radius}')
print(f'V(0)                : {-k/np.sqrt(eps):.4f}')
print(f'max force approx    : {2*k/(3*np.sqrt(3)*eps):.4f}')
````

## Diagnostic: following one ray

The exact number of phi jumps depends on eps, r, tspan, and n_steps. The qualitative distinction is robust: bound radial rays have turning points where kappa vanishes and phi jumps by about pi; scattering radial rays with L = 0 keep phi constant and kappa stays positive.


````python
def ray_diagnostics(x0=0.0, r=0.2, kappa=1.0, theta=0.0, tspan=(0, 1), n_steps=8000, plot=True):
    z0 = (x0 + r * np.cos(theta), r * np.sin(theta), kappa * np.cos(theta), kappa * np.sin(theta))
    tr = bicharacteristic_flow(op.symbol, z0, tspan, dim=2, method='symplectic', n_steps=n_steps)
    t = np.asarray(tr['t'])
    xi_ray = np.asarray(tr['xi'])
    eta_ray = np.asarray(tr['eta'])
    kap = np.hypot(xi_ray, eta_ray)
    phi = np.unwrap(np.arctan2(eta_ray, xi_ray))
    steps = np.diff(phi)
    jumps = np.where(np.abs(steps) > 1.0)[0]
    print(f'n_steps={n_steps}: kappa_min={kap.min():.4f}, {len(jumps)} jump(s) of phi')
    print(f'sizes/pi = {np.round(steps[jumps] / np.pi, 2).tolist()}')
    print(f'phi_final - phi_0 = {(phi[-1] - phi[0]) / np.pi:.2f} pi')
    if plot:
        fig, ax = plt.subplots(1, 2, figsize=(10, 3.2))
        ax[0].plot(t, kap)
        ax[0].set_xlabel('t')
        ax[0].set_ylabel('kappa')
        ax[1].plot(t, phi / np.pi)
        ax[1].set_xlabel('t')
        ax[1].set_ylabel('phi / pi')
        plt.tight_layout()
        plt.show()
````

````python
# Bound, source at the centre, r = core radius
ray_diagnostics(r=0.2, kappa=1.0, theta=0.0)
ray_diagnostics(r=0.2, kappa=1.0, theta=1.0)

# Scattering, source at the centre, r = core radius
ray_diagnostics(r=0.2, kappa=3.6, theta=0.0)

# Bound, deep inside the core
ray_diagnostics(r=0.02, kappa=1.0, theta=0.0)
````

## Choosing your own xi0

Use the central-energy relation p0 = kappa^2/2 - 1/sqrt(r^2 + eps) to choose xi0 = kappa for a desired central energy. The animations above use fixed kappa values for comparability; use this helper to design custom experiments.
