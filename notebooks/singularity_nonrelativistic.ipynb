# Ray fans through a softened Coulomb singularity

**Goal.** Follow the wavefront set of a *point source* of fixed energy $E$ as it is moved along the $x$-axis near the centre of an attractive Coulomb-type potential, in two regimes:

- **scattering** ($E>0$): the rays leave the source, pass the centre and escape to infinity; the soft core produces a *rainbow* (a fold caustic);
- **bound motion** ($E<0$): the rays stay in the potential well. For a pure Coulomb potential the whole fan would refocus on the source after one Kepler period; the soft core spoils this for the rays that cross it.

| Part | Content |
|---|---|
| A | scattering fans, four 2D slices of the phase space |
| B | bound fans and their refocusing after one Kepler period |
| C | the cotangent bundle itself: base plane + circle fibre, in 3D |
| D | diagnostic of a single ray |

## 1. The operator

We work in $\mathbb{R}^2$ with the symbol

$$
p(x,y,\xi,\eta)\;=\;\frac{\xi^{2}+\eta^{2}}{2m}\;-\;\frac{k}{\sqrt{x^{2}+y^{2}+\varepsilon}} .
$$

With the Kohn–Nirenberg quantization,

$$
(Pu)(x)=\frac{1}{(2\pi)^2}\iint e^{\,i x\cdot\xi}\,p(x,\xi)\,\hat u(\xi)\,d\xi
\quad\Longrightarrow\quad
P=-\frac{1}{2m}\Delta+V(x),\qquad V(x)=-\frac{k}{\sqrt{|x|^{2}+\varepsilon}}\qquad(\hbar=1),
$$

so $P$ is a Schrödinger operator with a **softened Coulomb potential**. Only the symbol `op.symbol` is used below.

| Parameter | Value | Meaning |
|---|---|---|
| $m$ | $1$ | mass |
| $k$ | $1$ | coupling (attractive) |
| $\varepsilon$ | $0.01$ | softening |

The singularity of the pure Coulomb symbol ($\varepsilon=0$, $V\sim -k/\lvert x\rvert$) at the origin is regularised by $\varepsilon$. The regularised potential has

- core radius $\sqrt{\varepsilon}=0.1$,
- depth $V(0)=-k/\sqrt{\varepsilon}=-10$,
- maximal force $F_{\max}=\dfrac{2k}{3\sqrt3\,\varepsilon}\approx 38.5$, reached at $\lvert x\rvert=\sqrt{\varepsilon/2}\approx 0.07$.

## 2. Rays = bicharacteristics

Singularities of a solution propagate along the Hamiltonian flow of $p$:

$$
\dot x=\frac{\xi}{m},\qquad
\dot y=\frac{\eta}{m},\qquad
\dot \xi=-\frac{k\,x}{(x^{2}+y^{2}+\varepsilon)^{3/2}},\qquad
\dot \eta=-\frac{k\,y}{(x^{2}+y^{2}+\varepsilon)^{3/2}} .
$$

Two quantities are conserved along each ray: the symbol itself, $p=E$ (the energy of the ray), and the angular momentum $L=x\eta-y\xi$ (the symbol is rotation invariant). The impact parameter with respect to the centre is $b=L/\kappa$, where $\kappa=\sqrt{\xi^2+\eta^2}$ is the covector norm.

> This is the **geometric (semiclassical) picture**: the animations show what the theory of propagation of singularities predicts. It is not a simulation of a wave equation (no amplitudes, no interference).

## 3. A point source on an energy shell

The singularities of a solution of $(P-E)u=f$ lie on the characteristic set $\{p=E\}$ and travel along its bicharacteristics. A point source at $(x_0,0)$ contains *every* direction of the covector, so it is modelled by $N$ seeds on a small circle of radius $r$ around the source, each carrying an outward covector **whose norm is chosen so that $p=E$**:

$$
(x_j,y_j)=\bigl(x_0+r\cos\theta_j,\; r\sin\theta_j\bigr),\qquad
(\xi_j,\eta_j)=\kappa_j\bigl(\cos\theta_j,\;\sin\theta_j\bigr),\qquad
\kappa_j=\sqrt{2m\,\bigl(E-V(x_j,y_j)\bigr)},\qquad
\theta_j=\frac{2\pi j}{N}.
$$

In `animate_wavefront_set` this is the option `energy=E` (it replaces the choice of `xi0`, `eta0`). A seed in the classically forbidden region ($V>E$) has no real $\kappa_j$ and is dropped.

The only effect of the radius $r$ is that every ray starts at a distance $r$ from the ideal point source, so the picture is shifted by about $r$ with respect to the point-source limit.

### Scattering energy $E=2$

For a plane wave of rays arriving from far away with impact parameter $b$, the deflection angle $\Theta(b)$ of a pure Coulomb potential is monotonic. The soft core makes it non-monotonic: it has a **maximum $\Theta_{\max}$**, and the rays pile up in a fold caustic at that angle (a rainbow). Computed numerically for this potential:

| $E$ | $0.5$ | $1$ | $2$ | $5$ | $10$ | $25$ |
|---|---|---|---|---|---|---|
| $\Theta_{\max}$ (degrees) | $93$ | $75$ | $56$ | $34$ | $21$ | $10$ |
| at $b\approx$ | $0.60$ | $0.40$ | $0.30$ | $0.20$ | $0.15$ | $0.15$ |

$E=2$ gives a large rainbow ($56^\circ$) at $b\approx0.3$, i.e. three core radii, and the rays are still fast enough to leave the core region quickly. For $E\ge10$ the fan is hardly deflected, for $E\le0.5$ the deflection reaches $90^\circ$ and the rays start to orbit the centre. In the source geometry the impact parameter of a ray is $b=x_0\sin\theta$, so a source at $x_0\gtrsim0.3$ covers the rainbow.

### Bound energy $E=-0.5$

For the Coulomb potential ($\varepsilon\to0$), orbits of energy $E$ are ellipses of semi-major axis $a_K=k/(2\lvert E\rvert)=1$ and period $T_K=2\pi\sqrt{m a_K^3/k}=2\pi$, **the same for all of them**. All the rays leaving a point therefore come back to it after exactly $T_K$: the fan refocuses. The classical turning radius is $2a_K=2$ ($V(r)=E$ at $r\approx1.99$), so the sources must satisfy $\lvert x_0\rvert<2$.

### Source radius

Median distance between the final positions of the same ray for a source of radius $r$ and for the reference $r=0.001$:

| $r$ | $0.5$ | $0.1$ | $0.02$ | $0.01$ |
|---|---|---|---|---|
| A: $E=2$, $x_0=1$, $T=3$ | $0.61$ | $0.13$ | $0.025$ | $0.012$ |
| B: $E=-0.5$, $x_0=0.5$, $T=2\pi$ | $0.51$ | $0.10$ | $0.020$ | $0.009$ |

The error is about $r$: $r=0.02$ is below $2\%$ of the size of the pictures, while $r=0.5$ is a ring-shaped source rather than a point source.

### Accuracy of the integration

Largest energy drift $\max\lvert p-E\rvert$ over the rays, on the worst source position of the sweep:

| | Verlet (the animations) | symplectic Euler (the previous version) |
|---|---|---|
| A: $E=2$, $T=3$, `n_steps=6000` | $5\cdot10^{-4}$ | about $10^{-2}$ |
| B: $E=-0.5$, $T=2\pi$, `n_steps=16000` | $2.5\cdot10^{-4}$ | about $3$–$5\cdot10^{-2}$ |

(The Euler values are the drift of the diagnostic of Part D, which uses `bicharacteristic_flow`.)

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import HTML

# Assuming psiop is on the python path
from psiop import *

# Spatial and frequency variables
x, y, xi, eta = sp.symbols('x y xi eta', real=True)

m = 1
k = 1
eps = 0.01

symbol_expr = (xi**2 + eta**2)/(2*m) - k/sp.sqrt(x**2 + y**2 + eps)

op = PseudoDifferentialOperator(
    expr=symbol_expr,
    vars_x=[x, y],
    mode='symbol',
    quantization='kohn-nirenberg',
    apply_backend='peetre',
)

print('Operator symbol:')
sp.pprint(op.symbol)
print(f'\ndimension           : {op.dim}')
print(f'quantization        : {op.quantization}')
print(f'apply backend       : {op.apply_backend}')
print(f'spatially dependent : {op._is_spatial_dependent()}')

# Energies of the two regimes and Kepler period of the bound one
E_scat = 2.0                                     # scattering
E_bound = -0.5                                   # bound
a_K = k/(2*abs(E_bound))                         # Kepler semi-major axis
T_K = 2*np.pi*np.sqrt(m*a_K**3/k)                # Kepler period
print(f'\nE_scat = {E_scat}, E_bound = {E_bound}, a_K = {a_K}, T_K = {T_K:.4f}')

plt.rcParams['animation.embed_limit'] = 2**128
```

# Part C: the cotangent bundle, base plane + circle fibre

The phase space $T^*\mathbb{R}^2=\mathbb{R}^2_{(x,y)}\times\mathbb{R}^2_{(\xi,\eta)}$ is a bundle over the base plane $(x,y)$; in polar coordinates of the fibre, $\xi=\kappa\cos\varphi$, $\eta=\kappa\sin\varphi$. With `coords3d=('x','y','phi')` and `color_by='kappa'` the plot shows

- the base $(x,y)$ as the horizontal plane,
- the covector angle $\varphi$ as the vertical axis (the circle fibre),
- the covector norm $\kappa$ as the colour.

At $t=0$ the green points form a (nearly) vertical line of height $2\pi$ above the source: the whole circle fibre, cut open at $\varphi=\pm\pi$. The flow carries it into a surface. The angle $\varphi$ is made continuous along each ray (only the $2\pi$ wraps are removed).

The covector turns only where the angular momentum is not zero:

$$
\dot\varphi=\frac{\xi\dot\eta-\eta\dot\xi}{\kappa^{2}}=\frac{V'(\rho)}{\rho\,\kappa^{2}}\,L,\qquad \rho=\sqrt{x^2+y^2}.
$$

- For a source at the centre ($L=0$) $\varphi$ stays constant along every ray and the scattering surface is a **helicoid**.
- At a turning point of a bound radial ray the covector vanishes and reverses, so $\varphi$ jumps by $\pi$ there (an artefact of the angle coordinate, not of the flow). For $x_0\neq0$, $L\neq0$ and the covector turns continuously, by about $\pi$ at each passage near the centre.

# Part D: diagnostic of a single ray

The function follows **one ray** of the point-source fan on the energy shell $p=E$ and reports how $\kappa$ and $\varphi$ evolve:

$$
\kappa=\sqrt{\xi^{2}+\eta^{2}},\qquad \varphi=\arg(\xi+i\eta).
$$

It prints the initial norm $\kappa_0$, the smallest value of $\kappa$, the energy drift $\max\lvert p-E\rvert$ along the ray, the number of **jumps** of $\varphi$ (steps larger than $1$ rad, i.e. of size about $\pi$) and the net change $\varphi_{\text{final}}-\varphi_0$, in units of $\pi$ and in degrees. The angle is computed with the plain `np.unwrap`, which removes every jump larger than $\pi$; the $(x,y,\varphi)$ animations of Part C remove only the $2\pi$ wraps. The ray is integrated with `bicharacteristic_flow` (symplectic Euler), so the energy drift is larger than in the animations (Section 4).

```python
def ray_diagnostics(op, energy, x0=0.0, r=0.02, theta=0.0, tspan=(0, 1), n_steps=8000, plot=True):
    '''Follow ONE ray of the point-source fan on the energy shell p = energy.'''
    p = sp.lambdify((x, y, xi, eta), op.symbol, 'numpy')
    X, Y = x0 + r*np.cos(theta), r*np.sin(theta)
    if p(X, Y, 0.0, 0.0) >= energy:
        raise ValueError('classically forbidden: p(x, y, 0, 0) >= energy at the seed')
    lo, hi = 0.0, 1.0                                         # covector norm: bisection on p = energy
    while p(X, Y, hi*np.cos(theta), hi*np.sin(theta)) < energy:
        hi *= 2
    for _ in range(80):
        mid = 0.5*(lo + hi)
        lo, hi = (mid, hi) if p(X, Y, mid*np.cos(theta), mid*np.sin(theta)) < energy else (lo, mid)
    kappa = 0.5*(lo + hi)
    z0 = (X, Y, kappa*np.cos(theta), kappa*np.sin(theta))
    tr = bicharacteristic_flow(op.symbol, z0, tspan, dim=2, method='symplectic', n_steps=n_steps)
    t, x_, y_, xi_, eta_ = (np.asarray(tr[c]) for c in ('t', 'x', 'y', 'xi', 'eta'))
    kap = np.hypot(xi_, eta_)
    phi = np.unwrap(np.arctan2(eta_, xi_))
    steps = np.diff(phi)
    jumps = np.where(np.abs(steps) > 1.0)[0]                  # steps of about +-pi
    drift = np.abs(p(x_, y_, xi_, eta_) - energy).max()
    print(f'kappa_0={kappa:.3f}, kappa_min={kap.min():.4f}, energy drift={drift:.1e}, '
          f'{len(jumps)} jump(s) of phi (sizes/pi = {np.round(steps[jumps]/np.pi, 2).tolist()}), '
          f'phi_final - phi_0 = {(phi[-1]-phi[0])/np.pi:.2f} pi ({np.degrees(phi[-1]-phi[0]):.0f} deg)')
    if plot:
        fig, ax = plt.subplots(1, 2, figsize=(10, 3.2))
        ax[0].plot(t, kap);       ax[0].set_xlabel('t'); ax[0].set_ylabel('kappa')
        ax[1].plot(t, phi/np.pi); ax[1].set_xlabel('t'); ax[1].set_ylabel('phi / pi')
        plt.tight_layout(); plt.show()

ray_diagnostics(op, E_bound, tspan=(0, T_K), n_steps=16000)                          # bound, from the centre
ray_diagnostics(op, E_bound, theta=1.0, tspan=(0, T_K), n_steps=16000)               # same, other direction
ray_diagnostics(op, E_scat, tspan=(0, 3), n_steps=8000)                              # scattering, from the centre
ray_diagnostics(op, E_scat, x0=1.0, theta=np.pi*1.05, tspan=(0, 3), n_steps=8000)    # scattering, deflected ray
```
